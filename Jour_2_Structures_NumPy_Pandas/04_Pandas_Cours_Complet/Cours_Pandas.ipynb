## 1. Qu'est-ce que Pandas et pourquoi l'utiliser ?

**Pandas** est la bibliothèque Python de référence pour manipuler des données
**tabulaires** — exactement comme un tableau Excel ou une table SQL : des lignes, des
colonnes nommées, et un index.

**Pandas est construit AU-DESSUS de NumPy** (`Cours_NumPy.ipynb`) : chaque colonne
Pandas est, en interne, un tableau NumPy. C'est pour cela que les calculs vectorisés
(sans boucle `for`) fonctionnent exactement pareil.

| | NumPy | Pandas |
|---|---|---|
| Structure | tableau de nombres, SANS nom de colonne | tableau AVEC noms de colonnes + index |
| Type de données | un seul type par tableau | un type différent par colonne (texte, date, nombre...) |
| Cas d'usage | calcul numérique pur | fichiers CSV/Excel/SQL, données métier réelles |
| Équivalent proche | un tableau de calcul brut | une feuille Excel ou une table SQL |

**Pourquoi c'est indispensable pour Beobank ?** Les fichiers source
(`CTR.csv`, `TIE.csv`, `TXN_X_CTR.csv`...) ont des colonnes nommées, des dates, du texte
ET des nombres mélangés dans le même fichier — exactement ce pour quoi Pandas a été conçu.
C'est l'équivalent Python de SAS (jeux de données / `PROC` ...) ou de SQL.

## 3. Series — une colonne étiquetée

Une **Series** Pandas est une seule colonne de données, avec un **index** (une étiquette)
pour chaque valeur. C'est la brique de base : un DataFrame (Module 4) est un assemblage
de plusieurs Series alignées sur le même index.

Par défaut, l'index est numéroté automatiquement à partir de 0 — mais on peut lui donner
des **étiquettes personnalisées**, par exemple les identifiants de compte.

```python
import pandas as pd

# --- Series avec index par défaut (0, 1, 2...) ---
soldes = pd.Series([15234.50, -500.0, 8900.0])   # pd.Series(liste) : construit une colonne étiquetée
print(soldes)
print(type(soldes))       # type() : indique la classe Python -> <class 'pandas.core.series.Series'>

# Accès par position, comme une liste
print("Élément à la position 1 :", soldes[1])    # [1] : accès par la position dans l'index (ici position = label)
```

```python
import pandas as pd
print(pd.DataFrame())   
```

```python
# --- Series avec index PERSONNALISÉ : les identifiants de compte comme étiquettes ---
soldes_par_compte = pd.Series(
    [15234.50, -500.0, 8900.0],                          # les VALEURS de la colonne
    index=["AC00001", "AC00002", "AC00003"]                # index= : une ÉTIQUETTE par valeur, dans le même ordre
)
print(soldes_par_compte)

# Accès par label (le nom de l'index), pas par position
print("Solde de AC00002 :", soldes_par_compte["AC00002"])   # ["AC00002"] : accès par LABEL d'index (pas par position)

# Une Series peut aussi se construire à partir d'un dictionnaire :
# les clés deviennent automatiquement l'index
depuis_dict = pd.Series({"AC00001": 15234.50, "AC00002": -500.0})   # pd.Series(dict) : clés -> index, valeurs -> données
print(depuis_dict)
```

```python
a = pd.Series([1, 2, 3, 4, 5])
print(a[2])
print(a[0])
```

## 4. DataFrames — un tableau étiqueté

Un **DataFrame** est un tableau à 2 dimensions : plusieurs colonnes (chacune est une
Series), toutes alignées sur le **même index** de lignes. C'est l'objet que l'on utilise
en pratique 95 % du temps avec Pandas — l'équivalent d'une feuille Excel ou d'une table SQL.

```python
import pandas as pd

# --- Construire un DataFrame à la main, depuis un dictionnaire ---
# pd.DataFrame(dict) : chaque CLÉ du dictionnaire devient un NOM DE COLONNE,
# chaque LISTE associée devient le CONTENU de la colonne (même longueur obligatoire)
mini_comptes = pd.DataFrame({
    "IDT_AC": ["AC001", "AC002", "AC003"],
    "SOLDE":  [1500.0, -200.0, 8300.0],
    "DEVISE": ["EUR", "EUR", "USD"],
})
print(mini_comptes)
print(type(mini_comptes))    # type() -> <class 'pandas.core.frame.DataFrame'>
```

```python
# --- .loc[] : localiser une ligne PAR SON LABEL D'INDEX ---
# Par défaut l'index est 0, 1, 2... -> .loc[1] donne la 2e ligne
print(mini_comptes.loc[1])              # .loc[label] : sélectionne la ligne dont l'index vaut 1

# .set_index("col") : remplace l'index numérique par le CONTENU d'une colonne existante
comptes_indexes = mini_comptes.set_index("IDT_AC")
print(comptes_indexes.loc["AC002"])       # localise directement par identifiant de compte, plus lisible
```

## 5. Lire un CSV (Read CSV)

`pd.read_csv()` charge un fichier CSV directement en DataFrame — plus simple et plus
rapide que `csv.DictReader` (vu au Jour 1), et le résultat est directement exploitable
avec toutes les méthodes Pandas.

**Les fichiers Beobank ont trois particularités** qu'il faut indiquer explicitement à
`read_csv()`, sinon le fichier est mal lu :

| Paramètre | Pourquoi | Valeur Beobank |
|---|---|---|
| `sep` | le séparateur entre colonnes n'est pas la virgule US standard | `";"` |
| `na_values` | indique quel texte doit être compris comme "valeur manquante" | `"."` |
| `encoding` | garantit que les caractères accentués (é, à...) s'affichent bien | `"utf-8"` |

```python
import pandas as pd

# pd.read_csv(chemin, ...) : lit le fichier CSV et construit un DataFrame
ctr = pd.read_csv(
    "../../data/CTR.csv",
    sep=";",           # sep=";" : le séparateur de colonnes du fichier est un point-virgule
    na_values=".",      # permet à Pandas de les identifier comme des données manquantes, 
                        #ce qui débloque les calculs mathématiques et simplifie le nettoyage des données.
    encoding="utf-8",   # encoding="utf-8" : garantit un affichage correct des caractères accentués
)
print(f"Dimensions : {ctr.shape[0]} lignes, {ctr.shape[1]} colonnes")
print(ctr.head(3))     # .head(n) : affiche les n premières lignes, pour un premier coup d'œil
print(ctr.tail(3))     # .tail(n) : affiche les n dernières lignes, pour un dernier coup d'œil
              # .shape : tuple (lignes, colonnes)
```

💡 Par défaut, Pandas **tronque** l'affichage d'un DataFrame trop large. Deux astuces

## 9. Cellules vides (Empty Cells)

Une valeur manquante s'appelle `NaN` (*Not a Number*) en Pandas. Trois méthodes clés :

- `.isna()` : renvoie `True`/`False` pour chaque cellule (souvent suivi de `.sum()`
  pour compter par colonne)
- `.dropna()` : **supprime** les lignes (ou colonnes) contenant des manquants
- `.fillna(valeur)` : **remplace** les manquants par une valeur (0, une moyenne, un texte...)

**Quelle stratégie choisir ?** Ça dépend du métier : supprimer une ligne perd de
l'information (à éviter si on peut compléter intelligemment), remplacer par 0 peut fausser
une moyenne. Il n'y a pas de règle universelle — c'est une décision d'analyse.

```python
ctr = pd.read_csv("../../data/CTR.csv", sep=";", na_values=".", encoding="utf-8")

# --- Détecter ---
print("Valeurs manquantes par colonne :")
print(ctr.isna().sum())              # .isna() : True/False par cellule ; .sum() sur des booléens = nombre de True, colonne par colonne

nb_manquants_solde = ctr["SLD_CTR"].isna().sum()    # même logique, restreinte à UNE colonne
print(f"\nSoldes manquants : {nb_manquants_solde} sur {len(ctr)}")
```

```python
# --- Stratégie 1 : supprimer les lignes concernées ---
ctr_sans_manquant = ctr.dropna(subset=["SLD_CTR"])     # .dropna(subset=[...]) : supprime les lignes manquantes SUR CES colonnes
print(f"Lignes conservées après dropna : {len(ctr_sans_manquant)} / {len(ctr)}")

# --- Stratégie 2 : remplacer par une valeur fixe ---
ctr_zero = ctr.copy()                                    # .copy() : travaille sur une copie, ne touche pas ctr original
ctr_zero["SLD_CTR"] = ctr_zero["SLD_CTR"].fillna(0)       # .fillna(0) : remplace chaque NaN par 0 (hypothèse "pas de solde connu = 0")
print(f"Manquants après fillna(0) : {ctr_zero['SLD_CTR'].isna().sum()}")

# --- Stratégie 3 : remplacer par la moyenne (garde la distribution globale) ---
ctr_moyenne = ctr.copy()
moyenne_solde = ctr_moyenne["SLD_CTR"].mean()               # .mean() : moyenne de la colonne (NaN ignorés automatiquement)
ctr_moyenne["SLD_CTR"] = ctr_moyenne["SLD_CTR"].fillna(moyenne_solde)   # remplace les NaN par cette moyenne
print(f"Moyenne utilisée pour remplacer : {moyenne_solde:.2f}")
```

## 10. Format incorrect (Wrong Format)

Une colonne peut contenir la BONNE information, mais dans le MAUVAIS type — le cas le
plus fréquent : une date stockée comme texte (`object`) au lieu d'un vrai type `datetime`.
Tant qu'elle reste du texte, impossible de calculer une durée, extraire une année, trier
chronologiquement de façon fiable...

**Les fichiers Beobank utilisent DEUX formats de date différents** :
- `CTR.csv`, `TIE.csv`, `TXN_X_CTR.csv` : format standard `AAAA-MM-JJ` (ISO)
- `TIE_ADR.csv` : format `JJMOISAAAA` (ex : `"24NOV2025"`) — il faut `format="mixed"`
  pour que Pandas devine automatiquement le bon format ligne par ligne.

```python
ctr = pd.read_csv("../../data/CTR.csv", sep=";", na_values=".", encoding="utf-8")

print("Type AVANT conversion :", ctr["DAT_OUV_CTR"].dtype)    # .dtype : object = texte, pas encore une vraie date

ctr["DAT_OUV_CTR"] = pd.to_datetime(ctr["DAT_OUV_CTR"])         # pd.to_datetime() : texte -> vrai type date (format ISO reconnu directement)

print("Type APRÈS conversion :", ctr["DAT_OUV_CTR"].dtype)    # datetime64[ns] : type date natif de Pandas

# Une fois au bon format, l'accesseur .dt donne accès à toutes les composantes
ctr["ANNEE_OUV"] = ctr["DAT_OUV_CTR"].dt.year     # .dt.year : extrait l'année de chaque date
ctr["MOIS_OUV"]  = ctr["DAT_OUV_CTR"].dt.month    # .dt.month : extrait le mois (1 à 12)
print(ctr[["IDT_AC", "DAT_OUV_CTR", "ANNEE_OUV", "MOIS_OUV"]].head(3))
```

```python
# --- Format mixte : TIE_ADR.csv utilise "24NOV2025" (jour-MOIS_TEXTE-année) ---
tie_adr = pd.read_csv("../../data/TIE_ADR.csv", sep=";", na_values=".", encoding="utf-8")

# format="mixed" : Pandas essaie de deviner automatiquement le format de chaque valeur
tie_adr["DAT_MAJ_ADR"] = pd.to_datetime(tie_adr["DAT_MAJ_ADR"], format="mixed")   # format="mixed" -> gère "24NOV2025"
print(tie_adr[["IDT_PI", "DAT_MAJ_ADR"]].head(3))

# errors="coerce" : si une valeur est vraiment impossible à convertir, la transformer
# en NaT (Not a Time, l'équivalent NaN pour les dates) plutôt que de planter le programme
dates_robustes = pd.to_datetime(tie_adr["DAT_MAJ_ADR"], format="mixed", errors="coerce")   # errors="coerce" : sécurise la conversion
print("Dates non convertibles :", dates_robustes.isna().sum())    # .isna() détecte aussi les NaT (dates manquantes)
```

## 11. Données incorrectes (Wrong Data)

Une valeur peut être du bon TYPE (un nombre reste un nombre) mais **métier-ment
impossible ou suspecte** : un solde extrêmement élevé qui ressemble à une erreur de
saisie, un âge négatif... Contrairement au format (section 10), ici la donnée EST un
nombre valide — c'est sa VALEUR qui pose question.

**Il n'y a pas de détection 100 % automatique** : on pose des règles métier (des seuils),
on inspecte, puis on décide (corriger, supprimer, ou juste flaguer pour vérification humaine).

```python
ctr = pd.read_csv("../../data/CTR.csv", sep=";", na_values=".", encoding="utf-8")

# --- Détecter les valeurs extrêmes avec des quantiles (plus robuste qu'un seuil fixe) ---
q_bas, q_haut = ctr["SLD_CTR"].quantile([0.01, 0.99])     # .quantile([0.01, 0.99]) : valeurs des 1er et 99e centiles
print(f"1% des soldes sont en dessous de {q_bas:.2f}")
print(f"1% des soldes sont au-dessus de {q_haut:.2f}")

suspects = ctr[(ctr["SLD_CTR"] < q_bas) | (ctr["SLD_CTR"] > q_haut)]   # filtre : en dehors de la plage normale (| = OU)
print(f"Comptes suspects (hors 1er-99e centile) : {len(suspects)}")
```

```python
# --- Corriger : on utilise .loc[condition, colonne] pour cibler et corriger PRÉCISÉMENT ---
# Exemple : plafonner (capper) les valeurs aberrantes plutôt que les supprimer
ctr_corrige = ctr.copy()                                                     # copie indépendante, pour ne pas modifier ctr
ctr_corrige.loc[ctr_corrige["SLD_CTR"] > q_haut, "SLD_CTR"] = q_haut          # .loc[condition, colonne] = valeur : modification CIBLÉE
print("Nouveau maximum après plafonnement :", ctr_corrige["SLD_CTR"].max())   # .max() : vérifie que plus rien ne dépasse q_haut

# Exemple : simplement FLAGUER pour vérification humaine, sans rien modifier
ctr["A_VERIFIER"] = (ctr["SLD_CTR"] < q_bas) | (ctr["SLD_CTR"] > q_haut)      # nouvelle colonne booléenne (True/False)
print(ctr["A_VERIFIER"].value_counts())    # .value_counts() : compte les True et les False
```

## 12. Doublons (Removing Duplicates)

- `.duplicated()` : renvoie `True` pour chaque ligne déjà vue avant (la 1ère occurrence
  reste `False`, les suivantes deviennent `True`)
- `.drop_duplicates()` : supprime les doublons (garde la 1ère occurrence par défaut)

**Attention au piège métier :** dans `TIE_X_CTR.csv`, un même `IDT_PI` (client) peut
apparaître PLUSIEURS FOIS **légitimement** — un client peut posséder plusieurs contrats !
Ce n'est PAS un doublon à supprimer. Le vrai doublon à traquer serait une ligne
**strictement identique sur toutes les colonnes**, signe d'une erreur d'import.

```python
txc = pd.read_csv("../../data/TIE_X_CTR.csv", sep=";", na_values=".", encoding="utf-8")

# --- Cas 1 : répétitions LÉGITIMES (un client, plusieurs contrats) ---
print(f"Lignes dans TIE_X_CTR       : {len(txc)}")
print(f"Clients (IDT_PI) uniques    : {txc['IDT_PI'].nunique()}")   # .nunique() : nombre de valeurs DISTINCTES

repetitions_client = txc["IDT_PI"].duplicated().sum()      # .duplicated() : True pour chaque répétition (pas la 1ère occurrence)
print(f"IDT_PI répétés (multi-contrats) : {repetitions_client}  <- normal, PAS un doublon à supprimer")

# --- Cas 2 : vrais doublons -> ligne ENTIÈREMENT identique à une autre ---
vrais_doublons = txc.duplicated().sum()      # .duplicated() SANS subset= -> compare TOUTES les colonnes à la fois
print(f"Lignes strictement dupliquées : {vrais_doublons}")

txc_propre = txc.drop_duplicates()            # .drop_duplicates() : supprime uniquement les vrais doublons (garde la 1ère occurrence)
print(f"Après drop_duplicates() : {len(txc_propre)} lignes")
```

## 13. Corrélations

`.corr()` calcule, pour chaque paire de colonnes numériques, un **coefficient de
corrélation** entre -1 et 1 :

- proche de **+1** : les deux colonnes augmentent ensemble (corrélation positive forte)
- proche de **-1** : quand l'une augmente, l'autre diminue (corrélation négative forte)
- proche de **0** : pas de relation linéaire visible entre les deux colonnes

⚠️ **Corrélation n'est PAS causalité.** Deux colonnes corrélées ne veulent pas dire que
l'une CAUSE l'autre — c'est un signal à interpréter avec le métier, jamais une preuve.

```python
ctr = pd.read_csv("../../data/CTR.csv", sep=";", na_values=".", encoding="utf-8")

# .corr() : calcule le coefficient de corrélation pour CHAQUE paire de colonnes numériques
correlations = ctr[["SLD_CTR", "MNT_INI", "SLD_DSP"]].corr()
print(correlations.round(2))     # .round(2) : arrondit l'affichage à 2 décimales

# Lire une seule paire de colonnes
print(f"\nCorrélation SLD_CTR / SLD_DSP : {correlations.loc['SLD_CTR', 'SLD_DSP']:.2f}")   # .loc[ligne, colonne] : une seule case du tableau
```

## 14. Graphiques avec Pandas (Plotting)

Pandas intègre Matplotlib en coulisses : la méthode `.plot()` permet de tracer un
graphique **directement depuis un DataFrame ou une Series**, en une ligne, sans passer
par `fig, ax = plt.subplots()`. Pratique pour une exploration rapide — pour un graphique
soigné destiné à une présentation, Jour 3 détaille l'usage complet de Matplotlib.

`kind=` choisit le type de graphique : `"line"` (défaut), `"bar"`, `"barh"`, `"hist"`,
`"scatter"`, `"pie"`...

```python
import matplotlib.pyplot as plt   # bibliothèque de graphiques utilisée en coulisses par .plot()

ctr = pd.read_csv("../../data/CTR.csv", sep=";", na_values=".", encoding="utf-8")

# --- Histogramme : distribution des soldes ---
# .plot(kind="hist", ...) : trace un histogramme directement depuis la colonne Series
ctr["SLD_CTR"].plot(kind="hist", bins=15, title="Distribution des soldes (SLD_CTR)")   # bins=15 : 15 barres de répartition
plt.xlabel("Solde (EUR)")     # plt.xlabel() : légende de l'axe horizontal
plt.show()                     # plt.show() : affiche le graphique
```

```python
# --- Barres : nombre de contrats par statut ---
ctr["COD_ECV_CTR"].value_counts().sort_index().plot(   # .value_counts() compte, .sort_index() trie par code, .plot() dessine
    kind="bar", title="Contrats par statut (COD_ECV_CTR)", color="#0E8C86"   # kind="bar" : barres verticales
)
plt.xlabel("Code statut")
plt.ylabel("Nombre de contrats")   # plt.ylabel() : légende de l'axe vertical
plt.show()
```

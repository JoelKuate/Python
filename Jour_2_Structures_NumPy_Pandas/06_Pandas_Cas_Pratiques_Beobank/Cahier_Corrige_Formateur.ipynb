## Exercice 2 — Préparer un mailing marketing

**Mise en situation :** Le service marketing veut envoyer un email à tous les clients. Il vous demande : « Est-ce qu'on a surtout des particuliers ou des entreprises ? Et dans quelle langue on doit écrire l'email ? »

**Étape 2.1.** `COD_TYP_TIE` est un code : 1 = particulier, 2 = entreprise. Utilisez **`.map()`** pour créer une colonne `TYPE_CLIENT` avec des libellés clairs.

```python
# COD_TYP_TIE est une colonne de tie : un CODE (1 ou 2) qui dit si le client est une personne
# physique (un particulier) ou une personne morale (une entreprise)
# .map({...}) remplace chaque code par le texte qu'on lui associe dans le dictionnaire
tie["TYPE_CLIENT"] = tie["COD_TYP_TIE"].map({1: "Particulier", 2: "Entreprise"})
```

**Étape 2.2.** Comptez combien de clients il y a dans chaque catégorie avec **`.value_counts()`**.

```python
# .value_counts() compte combien de fois chaque valeur différente apparaît dans la colonne
# Ici : combien de "Particulier" et combien d'"Entreprise" dans la colonne TYPE_CLIENT qu'on vient de créer
tie["TYPE_CLIENT"].value_counts()
```

**Étape 2.3.** Faites la même chose sur la colonne de la langue de contact : `COD_LNG_CTR`.

```python
# COD_LNG_CTR est la colonne de tie qui indique la langue de contact du client : "FR" ou "NL"
# On applique la même méthode .value_counts() pour compter chaque langue
tie["COD_LNG_CTR"].value_counts()
```

**Étape 2.4.** Comptez le nombre de particuliers avec **`.sum()`** sur un test d'égalité (un test comme `== "Particulier"` renvoie des `True`/`False`, et `.sum()` compte les `True`).

```python
# (tie["TYPE_CLIENT"] == "Particulier") compare chaque ligne à "Particulier"
# et renvoie une colonne de True (si c'est vrai) ou False (si c'est faux)
# .sum() additionne cette colonne : un True compte pour 1, un False compte pour 0
# -> le résultat est donc le nombre total de particuliers
nb_particuliers = (tie["TYPE_CLIENT"] == "Particulier").sum()
print(nb_particuliers)
```

💡 **Ce que ça veut dire pour la banque :** Le portefeuille est presque entièrement composé de particuliers, majoritairement francophones : le marketing peut écrire l'email en français, avec une version néerlandaise pour les autres.

## Exercice 3 — Nettoyer les comptes bancaires

**Mise en situation :** Un auditeur interne ouvre le fichier des comptes (`CTR.csv`) et se plaint : « Je ne peux rien calculer, tout est en texte ! Et il y a des points partout ! » Il faut nettoyer le fichier avant de pouvoir l'utiliser.

**Étape 3.1.** Chargez `CTR.csv` dans une variable `ctr` (même méthode qu'à l'exercice 1), puis affichez `.dtypes` pour constater que tout est en texte.

```python
# CTR.csv est le fichier des COMPTES BANCAIRES (aussi appelés contrats) de la banque
ctr = pd.read_csv("../../data/CTR.csv", sep=";", encoding="cp1252")

# .dtypes : on regarde le type de chaque colonne
# Résultat : tout est en texte (str), même les montants (SLD_CTR...) et les dates (DAT_OUV_CTR...)
# C'est normal : tant qu'on n'a rien nettoyé, Pandas ne sait pas encore que ce sont des nombres ou des dates
ctr.dtypes
```

**Étape 3.2.** Dans ce fichier, une valeur manquante est écrite `"."` (un simple point). Remplacez-la par une vraie valeur manquante Pandas avec **`.replace()`**.

```python
# Dans cet export bancaire, une case vide n'est pas vraiment vide : elle contient le texte "."
# .replace(".", pd.NA) remplace chaque "." par pd.NA, la vraie façon pour Pandas de dire
# "cette valeur est manquante" (indispensable pour pouvoir faire des calculs ensuite)
ctr = ctr.replace(".", pd.NA)
```

**Étape 3.3.** Convertissez les colonnes de montants en nombres avec **`pd.to_numeric()`**. N'oubliez pas `errors="coerce"` (sinon la conversion plante à la première valeur bizarre).

```python
# SLD_CTR = solde du compte (l'argent qu'il y a dessus)
# SLD_DSP = solde disponible (l'argent que le client peut vraiment dépenser tout de suite)
# MNT_INI = montant initial (l'argent mis sur le compte à son ouverture)
# Ces 3 colonnes sont encore du texte : on les transforme en vrais nombres décimaux
# errors="coerce" : si une valeur ne peut pas être transformée en nombre, on met une valeur manquante (NaN)
# plutôt que de faire planter tout le programme
for colonne in ["SLD_CTR", "SLD_DSP", "MNT_INI"]:
    ctr[colonne] = pd.to_numeric(ctr[colonne], errors="coerce")
```

**Étape 3.4.** Convertissez les colonnes de dates avec **`pd.to_datetime()`**. Les dates de ce fichier sont écrites au format ISO (année-mois-jour), donc `format="%Y-%m-%d"`.

```python
# DAT_OUV_CTR = date d'ouverture du compte
# DAT_CLO_CTR = date de clôture du compte (remplie seulement si le compte est fermé)
# pd.to_datetime(...) transforme le texte en vraie date, utilisable pour des calculs (durées, tris...)
# format="%Y-%m-%d" : on dit à Pandas que les dates sont écrites "année-mois-jour" (ex: 2024-05-29)
for colonne in ["DAT_OUV_CTR", "DAT_CLO_CTR"]:
    ctr[colonne] = pd.to_datetime(ctr[colonne], format="%Y-%m-%d", errors="coerce")
```

**Étape 3.5.** `COD_ECV_CTR` est un code : 4 = compte actif, 6 = compte clôturé. Créez la colonne `STATUT_COMPTE` avec `.map()` (le mot pour le code 4 est "Actif").

```python
# COD_ECV_CTR est le code d'état du compte : 4 veut dire "compte actif", 6 veut dire "compte clôturé"
# .map({...}) remplace chaque code par un texte clair, plus facile à lire qu'un chiffre
ctr["STATUT_COMPTE"] = ctr["COD_ECV_CTR"].map({4: "Actif", 6: "Clôturé"})

# on vérifie le résultat avec .value_counts() : combien de comptes actifs, combien de clôturés
ctr["STATUT_COMPTE"].value_counts()
```

💡 **Ce que ça veut dire pour la banque :** Une fois nettoyé et typé, `ctr` permet enfin de faire des calculs sur les montants et de comparer des dates. Un peu plus de la moitié des comptes sont encore actifs.

## Exercice 4 — Repérer les comptes en découvert

**Mise en situation :** Le service risque prépare un comité de crédit demain matin. Il a besoin de la liste des comptes dont le solde est négatif (un « découvert » : le client a dépensé plus d'argent qu'il n'en avait).

**Étape 4.1.** Créez `comptes_decouvert` : gardez les lignes de `ctr` où `SLD_CTR` est strictement inférieur à 0 (utilisez le symbole `<`).

```python
# SLD_CTR = solde du compte. Un solde négatif veut dire que le compte est "en découvert"
# ctr["SLD_CTR"] < 0 renvoie une colonne de True/False (True si le solde est négatif sur cette ligne)
# ctr[...] avec cette colonne de True/False à l'intérieur ne garde que les lignes où c'est True
comptes_decouvert = ctr[ctr["SLD_CTR"] < 0]
print(comptes_decouvert.shape[0])
```

**Étape 4.2.** Créez une colonne `EN_DECOUVERT` avec **`np.where()`** : "Oui" si `SLD_CTR < 0`, sinon "Non".

```python
# np.where(condition, valeur_si_vrai, valeur_si_faux) : un "SI... ALORS... SINON..." qui s'applique
# d'un coup à toute la colonne (beaucoup plus rapide qu'une boucle ligne par ligne)
ctr["EN_DECOUVERT"] = np.where(ctr["SLD_CTR"] < 0, "Oui", "Non")

ctr["EN_DECOUVERT"].value_counts()
```

**Étape 4.3.** Triez `comptes_decouvert` du découvert le plus important au moins important avec **`.sort_values()`** (tri croissant par défaut : les plus négatifs en premier).

```python
# .sort_values("SLD_CTR") trie le tableau du plus petit au plus grand solde (tri croissant par défaut)
# Comme les découverts sont négatifs, le tri croissant met le PIRE découvert en premier
# .head(5) : on ne garde que les 5 premières lignes après le tri
comptes_decouvert.sort_values("SLD_CTR").head(5)
```

**Étape 4.4.** Calculez le montant total en découvert avec **`.sum()`**.

```python
# .sum() additionne toutes les valeurs de la colonne SLD_CTR des comptes en découvert
# Comme ce sont des soldes négatifs, le résultat est un grand nombre négatif : le trou total
total_decouvert = comptes_decouvert["SLD_CTR"].sum()
print(f"{total_decouvert:.2f} EUR")
```

💡 **Ce que ça veut dire pour la banque :** Le service risque a maintenant une liste précise des comptes à surveiller, triée du cas le plus grave au moins grave, prête pour la réunion.

## Exercice 5 — Protéger les clients âgés

**Mise en situation :** Le service conformité doit surveiller particulièrement les clients de 75 ans et plus (une règle pour protéger les personnes vulnérables). Il vous demande la liste de ces clients.

**Étape 5.1.** Convertissez `DAT_NAI` (date de naissance) en vraie date, avec **`pd.to_datetime()`** (format ISO).

```python
# DAT_NAI = date de naissance du client, dans la table tie
# On la transforme en vraie date pour pouvoir calculer un âge ensuite
tie["DAT_NAI"] = pd.to_datetime(tie["DAT_NAI"], format="%Y-%m-%d", errors="coerce")
```

**Étape 5.2.** Calculez l'âge : (date d'aujourd'hui moins date de naissance) donne un nombre de jours (`.dt.days`) ; divisez par **365.25** pour obtenir des années.

```python
# pd.Timestamp.today() donne la date d'aujourd'hui
aujourdhui = pd.Timestamp.today()

# (aujourdhui - tie["DAT_NAI"]) calcule, pour chaque client, une DURÉE entre sa naissance et aujourd'hui
# .dt.days transforme cette durée en nombre de jours
# on divise par 365.25 (et pas 365) pour tenir compte des années bissextiles -> on obtient l'âge en années
# .round(1) arrondit à une décimale, pour un résultat plus lisible
tie["AGE"] = ((aujourdhui - tie["DAT_NAI"]).dt.days / 365.25).round(1)
```

**Étape 5.3.** Découpez l'âge en tranches avec **`pd.cut()`**.

```python
# on définit les limites des tranches d'âge, et le nom de chaque tranche
bornes = [0, 25, 40, 60, 75, 120]
etiquettes = ["<25 ans", "25-39 ans", "40-59 ans", "60-74 ans", "75 ans et +"]

# pd.cut(colonne, bins=..., labels=...) transforme un nombre (l'âge) en catégorie (la tranche d'âge)
tie["TRANCHE_AGE"] = pd.cut(tie["AGE"], bins=bornes, labels=etiquettes)
```

**Étape 5.4.** Filtrez `tie` pour ne garder que les clients de 75 ans et plus (symbole « supérieur ou égal » : `>=`).

```python
# tie["AGE"] >= 75 renvoie une colonne de True/False (True si le client a 75 ans ou plus)
# tie[...] ne garde que les lignes où c'est True
clients_a_surveiller = tie[tie["AGE"] >= 75]
print(clients_a_surveiller.shape[0])
```

💡 **Ce que ça veut dire pour la banque :** La conformité dispose d'une liste exacte des clients de 75 ans et plus, pour mettre en place le suivi renforcé exigé par la réglementation.

## Exercice 6 — Vérifier les coordonnées des clients

**Mise en situation :** Avant une grosse campagne email, le service marketing veut être sûr d'une chose : « Est-ce qu'on a bien les coordonnées de tout le monde ? »

**Étape 6.1.** Chargez `TIE_ADR.csv` dans une variable `adr` (même méthode que d'habitude).

```python
# TIE_ADR.csv contient les ADRESSES et coordonnées des clients (email, téléphone, ville...)
adr = pd.read_csv("../../data/TIE_ADR.csv", sep=";", encoding="cp1252")
print(adr.shape)
```

**Étape 6.2.** Calculez le taux de complétude (%) de l'email avec **`.notna()`** puis **`.mean()`** (un `.mean()` sur des True/False donne un pourcentage, entre 0 et 1).

```python
# ADR_EMA = adresse email du client, dans la table adr
# .notna() renvoie True si la case n'est PAS vide, False si elle est manquante
# .mean() sur une colonne de True/False donne la PROPORTION de True (entre 0 et 1)
# on multiplie par 100 pour avoir un pourcentage
taux_email = adr["ADR_EMA"].notna().mean() * 100
print(f"{taux_email:.1f} %")
```

**Étape 6.3.** Trouvez les clients sans email NI téléphone : combinez trois conditions `.isna()` avec le symbole **`&`** (« et »).

```python
# ADR_EMA = email, NUM_TEL_MOB_INL = téléphone mobile, NUM_TEL_DOM_INL = téléphone fixe
# .isna() renvoie True si la case EST manquante
# & veut dire "ET" : on garde seulement les lignes où LES TROIS conditions sont vraies en même temps
injoignables = adr[
    adr["ADR_EMA"].isna() & adr["NUM_TEL_MOB_INL"].isna() & adr["NUM_TEL_DOM_INL"].isna()
]
print(injoignables.shape[0])
```

**Étape 6.4.** Extrayez le domaine email (ce qui suit le `@`) avec **`.str.split("@")`** puis `.str[-1]` (dernier élément).

```python
# .str.split("@") coupe chaque email en deux morceaux, à l'endroit du @
# (exemple : "jean@gmail.com" devient la liste ["jean", "gmail.com"])
# .str[-1] récupère le DERNIER élément de cette liste : le domaine (gmail.com, hotmail.com...)
adr["DOMAINE_EMAIL"] = adr["ADR_EMA"].str.split("@").str[-1]

# .value_counts() compte les domaines les plus fréquents, .head(5) garde les 5 premiers
adr["DOMAINE_EMAIL"].value_counts().head(5)
```

💡 **Ce que ça veut dire pour la banque :** La majorité des clients ont un email connu, mais certains sont totalement injoignables : pour eux, la campagne email ne fonctionnera pas.

## Exercice 7 — Enquêter sur un client qui a beaucoup de comptes

**Mise en situation :** Le directeur régional fronce les sourcils devant un tableau de bord : « Il y a un client qui a l'air d'avoir énormément de comptes. Vérifie ça ! »

**Étape 7.1.** Chargez `TIE_X_CTR.csv` dans une variable `lnk` (la table qui dit quel client possède quel compte).

```python
# TIE_X_CTR.csv est la table de LIAISON entre les clients et les comptes :
# chaque ligne dit "ce client (NUM_TIE) possède ce compte (IDT_AC)"
lnk = pd.read_csv("../../data/TIE_X_CTR.csv", sep=";", encoding="cp1252")
print(lnk.shape)
```

**Étape 7.2.** Comptez le nombre de comptes par client avec **`.groupby()`** puis `.size()`.

```python
# NUM_TIE = numéro du client. .groupby("NUM_TIE") regroupe toutes les lignes d'un même client ensemble
# .size() compte combien de lignes (donc combien de comptes) il y a dans chaque groupe
# .sort_values(ascending=False) trie du client avec le plus de comptes au client avec le moins
nb_comptes_par_client = lnk.groupby("NUM_TIE").size().sort_values(ascending=False)
nb_comptes_par_client.head()
```

**Étape 7.3.** Trouvez le client qui a le plus de comptes avec **`.idxmax()`** (donne l'étiquette de la plus grande valeur).

```python
# .idxmax() renvoie l'étiquette (ici le NUM_TIE) qui correspond à la plus grande valeur de la série
# .max() renvoie directement cette plus grande valeur (le nombre de comptes)
client_suspect = nb_comptes_par_client.idxmax()
print(client_suspect, nb_comptes_par_client.max())
```

**Étape 7.4.** Trouvez les clients de `tie` qui n'ont AUCUN compte dans `lnk`. Utilisez **`.isin()`** (est-ce que la valeur est dans la liste ?) précédé du symbole `~` (qui veut dire « non »).

```python
# tie["NUM_TIE"].isin(lnk["NUM_TIE"]) renvoie True pour chaque client de tie dont le numéro
# se retrouve AUSSI dans lnk (donc un client qui a bien un compte)
# ~ (le tilde) inverse le résultat : on garde alors les clients qui N'ONT PAS de compte dans lnk
clients_sans_compte = tie[~tie["NUM_TIE"].isin(lnk["NUM_TIE"])]
print(clients_sans_compte.shape[0])
```

💡 **Ce que ça veut dire pour la banque :** Un seul client concentre une grande partie des comptes de l'extrait : à vérifier manuellement. On voit aussi que beaucoup de clients de `tie` n'ont pas de compte dans `lnk` : deux exports qui ne couvrent pas exactement le même périmètre.

## Exercice 8 — Répondre à un client qui conteste des frais

**Mise en situation :** Un client appelle, énervé : « Il y a plein de prélèvements sur mon compte que je ne comprends pas ! » Son compte est le `65500477817`.

**Étape 8.1.** Chargez `TXN_X_CTR.csv` dans `txn`, puis convertissez `DAT_CRE_MVT_CPB` en date (format ISO).

```python
# TXN_X_CTR.csv contient les OPÉRATIONS (mouvements) faites sur les comptes bancaires
txn = pd.read_csv("../../data/TXN_X_CTR.csv", sep=";", encoding="cp1252")

# DAT_CRE_MVT_CPB = date à laquelle l'opération a été enregistrée
# on la transforme en vraie date pour pouvoir trier et comparer des dates ensuite
txn["DAT_CRE_MVT_CPB"] = pd.to_datetime(txn["DAT_CRE_MVT_CPB"], format="%Y-%m-%d", errors="coerce")
```

**Étape 8.2.** Filtrez `txn` pour ne garder que les opérations du compte `65500477817`, puis triez-les par date avec **`.sort_values()`**.

```python
# IDT_AC = identifiant du compte bancaire concerné par chaque opération
mon_compte = 65500477817

# txn["IDT_AC"] == mon_compte : True uniquement pour les lignes de ce compte précis
# .sort_values("DAT_CRE_MVT_CPB") trie les opérations trouvées de la plus ancienne à la plus récente
operations_client = txn[txn["IDT_AC"] == mon_compte].sort_values("DAT_CRE_MVT_CPB")
print(operations_client.shape[0])
```

**Étape 8.3.** Cherchez, parmi ces opérations, celles qui parlent de frais : utilisez **`.str.contains()`** sur `LIB_OPE_INL_1`, avec le motif `"frais|intér|forfait"` (le `|` veut dire « ou »).

```python
# LIB_OPE_INL_1 = libellé (texte qui décrit) l'opération bancaire, par exemple "Virement de" ou "Frais de rappel"
# .str.contains("frais|intér|forfait", case=False, na=False) cherche si l'un de ces trois mots
# apparaît dans le texte (le | veut dire "ou"). case=False : on ignore majuscules/minuscules.
# na=False : si le libellé est manquant, on considère que ce n'est pas un frais (pas d'erreur)
frais_du_client = operations_client[
    operations_client["LIB_OPE_INL_1"].str.contains("frais|intér|forfait", case=False, na=False)
]
print(frais_du_client.shape[0])
```

**Étape 8.4.** Affichez la liste des libellés de ces frais avec **`.value_counts()`**.

```python
# .value_counts() compte combien de fois chaque libellé de frais apparaît, pour ce client
frais_du_client["LIB_OPE_INL_1"].value_counts()
```

## Exercice 10 — Préparer une fiche client avant un rendez-vous

**Mise en situation :** Un conseiller a un rendez-vous avec un client important dans 10 minutes. Il vous demande une fiche rapide : tous ses comptes et son solde total.

**Étape 10.1.** Fusionnez `tie`, `lnk` et `ctr` en une seule table `vue_360` avec **`.merge()`**, en chaînant deux fusions (sur `NUM_TIE` puis sur `IDT_AC`).

```python
# .merge(autre_table, on="colonne_commune") rapproche deux tables qui ont une colonne en commun,
# comme une jointure en SQL. On enchaîne deux fusions pour tout assembler en une seule table :
# 1) tie (les clients) + lnk (qui possède quel compte), rapprochés sur NUM_TIE (numéro de client)
# 2) le résultat + ctr (les infos des comptes), rapproché sur IDT_AC (identifiant de compte)
# how="inner"/"left" précise quelles lignes garder si une correspondance manque d'un côté
vue_360 = tie.merge(lnk, on="NUM_TIE", how="inner").merge(ctr, on="IDT_AC", how="left")
vue_360.shape
```

**Étape 10.2.** Complétez la fonction `fiche_client` : le solde total se calcule avec **`.sum()`** sur la colonne `SLD_CTR`.

```python
def fiche_client(num_tie):
    # on garde uniquement les lignes de vue_360 qui correspondent à ce client précis
    comptes_du_client = vue_360[vue_360["NUM_TIE"] == num_tie]
    # SLD_CTR = solde de chaque compte. .sum() additionne les soldes de tous les comptes du client
    solde_total = comptes_du_client["SLD_CTR"].sum()
    # .nunique() compte le nombre de comptes DIFFÉRENTS (IDT_AC) que possède ce client
    nb_comptes = comptes_du_client["IDT_AC"].nunique()
    return f"Client {num_tie} : {nb_comptes} comptes, solde total = {solde_total:.2f} EUR"
```

**Étape 10.3.** Appelez `fiche_client()` pour le client `2900000004654` (le client repéré à l'exercice 7).

```python
# on appelle notre fonction avec le numéro du client repéré à l'exercice 7 (celui qui a 46 comptes)
print(fiche_client(2900000004654))
```

💡 **Ce que ça veut dire pour la banque :** En une fonction réutilisable, le conseiller obtient une phrase claire à lire juste avant d'entrer en rendez-vous, pour n'importe quel client.

## Exercice 12 — Préparer un audit qualité

**Mise en situation :** Un organisme de contrôle externe arrive la semaine prochaine. Le responsable conformité veut un petit rapport qualité des données.

**Étape 12.1.** Pour chaque table, comptez les lignes dupliquées avec **`.duplicated()`** puis `.sum()`.

```python
# on rassemble les 5 tables dans un dictionnaire, pour pouvoir les parcourir facilement
tables = {"ctr": ctr, "tie": tie, "adr": adr, "lnk": lnk, "txn": txn}

# .duplicated() renvoie True pour chaque ligne strictement identique à une ligne déjà vue avant elle
# .sum() compte le nombre de True, donc le nombre de lignes en double
for nom, table in tables.items():
    print(nom, table.duplicated().sum())
```

**Étape 12.2.** Vérifiez que la clé `IDT_AC` n'est jamais dupliquée dans `ctr` (comptez avec `.duplicated()` puis **`.sum()`**).

```python
# IDT_AC est censé identifier un compte de façon unique (une clé). Il ne devrait donc JAMAIS être dupliqué
doublons_cle = ctr["IDT_AC"].duplicated().sum()
print(doublons_cle)
```

**Étape 12.3.** Vérifiez qu'aucun compte n'est clôturé AVANT d'avoir été ouvert (comparez `DAT_CLO_CTR` et `DAT_OUV_CTR` avec `<`).

```python
# DAT_CLO_CTR = date de clôture, DAT_OUV_CTR = date d'ouverture
# Une date de clôture avant la date d'ouverture n'a aucun sens : c'est une incohérence à détecter
comptes_incoherents = ctr[ctr["DAT_CLO_CTR"] < ctr["DAT_OUV_CTR"]]
print(comptes_incoherents.shape[0])
```

**Étape 12.4.** Complétez la fonction `rapport_qualite` : le taux de complétude se calcule à partir de **`.isna()`** (1 moins la moyenne des valeurs manquantes).

```python
def rapport_qualite(table, nom):
    return {
        "table": nom,
        "nb_lignes": table.shape[0],
        "doublons": table.duplicated().sum(),
        # .isna() dit si chaque case est manquante (True/False), .mean() calcule la proportion
        # de cases manquantes sur tout le tableau ; 1 moins cette proportion = le taux de complétude
        "taux_completude_pct": round((1 - table.isna().mean().mean()) * 100, 1),
    }

# on applique cette fonction aux 5 tables, puis on rassemble les résultats dans un DataFrame
pd.DataFrame([rapport_qualite(t, nom) for nom, t in tables.items()])
```

💡 **Ce que ça veut dire pour la banque :** Ce rapport peut être montré directement à l'auditeur externe : nombre de doublons, cohérence des dates, taux de complétude par table.

## Exercice 13 — Vérifier des soldes suspects

**Mise en situation :** Un stagiaire n'est pas sûr de son export : « Je crois qu'il y a des soldes bizarres dans mon fichier, tu peux vérifier ? »

**Étape 13.1.** Calculez le premier quart (`q1`) et le troisième quart (`q3`) des soldes avec **`.quantile()`** (0.25 et 0.75).

```python
# SLD_CTR = solde du compte. .quantile(0.25) donne la valeur en dessous de laquelle se trouvent
# 25% des soldes ; .quantile(0.75) donne la valeur en dessous de laquelle se trouvent 75% des soldes
q1 = ctr["SLD_CTR"].quantile(0.25)
q3 = ctr["SLD_CTR"].quantile(0.75)
# l'écart entre les deux s'appelle l'écart interquartile (IQR), une mesure classique de dispersion
ecart = q3 - q1
```

**Étape 13.2.** Calculez les bornes normales, puis filtrez les soldes en dehors de ces bornes. Combinez les deux conditions avec **`|`** (« ou »).

```python
# on définit une zone "normale" autour de q1 et q3, 1.5 fois l'écart interquartile de chaque côté
borne_basse = q1 - 1.5 * ecart
borne_haute = q3 + 1.5 * ecart

# | veut dire "ou" : on garde les soldes qui sont EN DESSOUS de la borne basse
# OU AU-DESSUS de la borne haute -> les valeurs jugées "hors norme" par cette méthode statistique
valeurs_etranges = ctr[(ctr["SLD_CTR"] < borne_basse) | (ctr["SLD_CTR"] > borne_haute)]
print(valeurs_etranges.shape[0])
```

**Étape 13.3.** Affichez un résumé statistique complet des soldes avec **`.describe()`**.

```python
# .describe() donne en une seule fois : le nombre de valeurs, la moyenne, l'écart-type,
# le minimum, le maximum et les quartiles -> un bon réflexe avant de conclure trop vite
ctr["SLD_CTR"].describe()
```

💡 **Ce que ça veut dire pour la banque :** Attention : avec beaucoup de soldes à 0 ou manquants, la méthode statistique signale presque tout solde non nul comme « anormal ». Une vraie vérification demande toujours d'en discuter avec le métier.

```python
# .nlargest(3, "SLD_CTR") garde les 3 comptes avec le plus gros solde, pour chaque devise séparément
top_eur = ctr[ctr["COD_DEV"] == "EUR"].nlargest(3, "SLD_CTR")
top_usd = ctr[ctr["COD_DEV"] == "USD"].nlargest(3, "SLD_CTR")

# pd.concat([...]) empile plusieurs tableaux les uns SOUS les autres (contrairement à merge,
# qui les met côte à côte) : ici, on obtient les 6 comptes dans un seul tableau
top_par_devise = pd.concat([top_eur, top_usd])
```

**Étape 15.2.** Complétez le résumé final : le nombre de comptes en découvert se calcule avec un test `< 0` suivi de **`.sum()`**.

```python
# pd.Series({...}) construit un petit tableau à une colonne, une ligne par indicateur clé (un "KPI")
resume_final = pd.Series({
    "Nombre de clients": tie.shape[0],
    "Nombre de comptes": ctr.shape[0],
    "Comptes actifs": (ctr["STATUT_COMPTE"] == "Actif").sum(),
    # SLD_CTR < 0 : le compte est en découvert. .sum() compte combien de comptes sont dans ce cas
    "Comptes en découvert": (ctr["SLD_CTR"] < 0).sum(),
    "Nombre d'opérations": txn.shape[0],
})
resume_final
```

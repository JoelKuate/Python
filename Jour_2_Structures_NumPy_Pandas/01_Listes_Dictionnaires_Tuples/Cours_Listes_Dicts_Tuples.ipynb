## 1. Qu'est-ce qu'une liste ?

Une **liste** Python est une collection de valeurs :
- **ordonnée** : les éléments gardent toujours le même ordre, tant qu'on ne les modifie pas
- **modifiable** (*mutable*) : on peut ajouter, retirer, changer des éléments après création
- **autorise les doublons** : la même valeur peut apparaître plusieurs fois
- peut contenir des types **mélangés**, même si en pratique on utilise plutôt un seul type par liste

On l'écrit entre crochets `[ ]`, éléments séparés par des virgules.

**Pourquoi c'est utile pour Beobank ?** Une liste de comptes, une liste de soldes, une
liste de devises observées dans le portefeuille — c'est la structure la plus naturelle
pour représenter "plusieurs valeurs du même genre, dans un ordre donné".

```python
# list() : la fonction native pour CRÉER une liste vide, équivalent à []
comptes_vide = []
print(comptes_vide, type(comptes_vide))   # type() confirme la classe : <class 'list'>

# Liste de comptes Beobank (des chaînes de caractères)
comptes = ["AC00001", "AC00002", "AC00003", "AC00004"]
print(comptes)

# Liste de soldes (des nombres) -- même principe, un autre type
soldes = [15234.50, -500.0, 8900.0, 0.0]
print(soldes)

# len() : fonction native qui donne le NOMBRE d'éléments d'une liste
print("Nombre de comptes :", len(comptes))

# "in" : teste si une valeur EST PRÉSENTE dans la liste -> renvoie True/False
print("AC00002" in comptes)      # True
print("AC09999" in comptes)      # False
```

## 2. Accéder aux éléments (Access List Items)

Comme pour les tableaux NumPy (`Cours_NumPy.ipynb`) : **le premier élément a l'index 0**.
On peut aussi utiliser des index **négatifs** (compter depuis la fin) et le **slicing**
(`[debut:fin]`) pour extraire une plage d'éléments.

```python
comptes = ["AC00001", "AC00002", "AC00003", "AC00004", "AC00005"]

print(comptes[0])       # [0] : PREMIER élément -> "AC00001"
print(comptes[2])       # [2] : 3e élément (index 0,1,2) -> "AC00003"
print(comptes[-1])      # [-1] : DERNIER élément (index négatif) -> "AC00005"
print(comptes[-2])      # [-2] : avant-dernier élément -> "AC00004"

# --- Slicing : [debut:fin] -> debut INCLUS, fin EXCLU ---
print(comptes[1:3])     # index 1 et 2 (3 exclu) -> ["AC00002", "AC00003"]
print(comptes[:2])      # depuis le début jusqu'à l'index 2 exclu -> les 2 premiers
print(comptes[3:])      # depuis l'index 3 jusqu'à la fin
print(comptes[-2:])     # les 2 DERNIERS éléments (très utilisé : "les N derniers")

# "in" reste utilisable pour vérifier une présence avant d'accéder
if "AC00003" in comptes:
    print("Ce compte existe dans la liste.")
```

On modifie un élément en lui réassignant une valeur à son index. On peut aussi
remplacer une PLAGE entière d'éléments avec le slicing à gauche du `=`.

```python
soldes = [15234.50, -500.0, 8900.0, 0.0]

soldes[1] = -450.0             # [index] = nouvelle_valeur : remplace UN élément
print(soldes)

soldes[0:2] = [16000.0, 100.0]   # slicing à gauche du = : remplace TOUTE la plage d'un coup
print(soldes)

# Remplacer une plage par PLUS ou MOINS d'éléments que l'original : la liste change de taille
soldes[1:3] = [1.0, 2.0, 3.0]    # remplace 2 éléments par 3 -> la liste s'agrandit
print(soldes)
```

## 4. Ajouter des éléments (Add List Items)

Trois méthodes pour ajouter des éléments — chacune a un rôle précis :

| Méthode | Ajoute | Où |
|---|---|---|
| `.append(x)` | **UN** élément `x` | à la fin |
| `.extend(iterable)` | **PLUSIEURS** éléments (d'une autre liste, tuple...) | à la fin, un par un |
| `.insert(i, x)` | **UN** élément `x` | à la position `i` précise |

```python
comptes = ["AC00001", "AC00002"]

comptes.append("AC00003")             # .append(x) : ajoute UN SEUL élément à la fin
print(comptes)

nouveaux_comptes = ["AC00004", "AC00005"]
comptes.extend(nouveaux_comptes)      # .extend(liste) : ajoute CHAQUE élément de l'autre liste, à la fin
print(comptes)

comptes.insert(0, "AC00000")          # .insert(position, valeur) : insère À la position donnée (décale les suivants)
print(comptes)

# Piège fréquent : append() avec une liste ajoute la LISTE ENTIÈRE comme UN SEUL élément (pas ce qu'on veut, en général)
demo = ["a", "b"]
demo.append(["c", "d"])                # ajoute la liste ["c","d"] comme un seul bloc imbriqué
print("append(liste) :", demo)          # ['a', 'b', ['c', 'd']]  <- probablement pas l'effet voulu
```

## 5. Supprimer des éléments (Remove List Items)

| Méthode / instruction | Supprime | Par |
|---|---|---|
| `.remove(x)` | la 1ère occurrence de la VALEUR `x` | valeur |
| `.pop(i)` | l'élément à la position `i` (et le RENVOIE) — sans `i`, retire le dernier | position |
| `del liste[i]` | l'élément à la position `i` (instruction du langage, pas une méthode) | position |
| `.clear()` | TOUS les éléments — la liste devient vide `[]` | — |

```python
comptes = ["AC00001", "AC00002", "AC00003", "AC00004", "AC00005"]

comptes.remove("AC00002")        # .remove(valeur) : cherche puis supprime la 1ère occurrence de cette VALEUR
print("Après remove :", comptes)

dernier = comptes.pop()          # .pop() SANS argument : retire et RENVOIE le DERNIER élément
print("Élément retiré par pop() :", dernier)
print("Après pop() :", comptes)

premier = comptes.pop(0)         # .pop(0) : retire et renvoie l'élément à la POSITION 0
print("Élément retiré par pop(0) :", premier)
print("Après pop(0) :", comptes)

del comptes[0]                   # del liste[i] : instruction Python (pas une méthode) -> supprime par position, sans rien renvoyer
print("Après del :", comptes)

comptes.clear()                  # .clear() : vide complètement la liste
print("Après clear() :", comptes)
```

## 6. Parcourir une liste (Loop Lists)

Plusieurs façons de boucler sur une liste — chacune a son usage :

```python
comptes = ["AC00001", "AC00002", "AC00003"]

# --- Méthode 1 : la plus courante -> on ne se soucie pas de la position ---
for compte in comptes:            # for VALEUR in liste : parcourt directement les VALEURS
    print(compte)

print("---")

# --- Méthode 2 : quand on a besoin de la POSITION en plus de la valeur ---
for position, compte in enumerate(comptes):     # enumerate(liste) : donne (position, valeur) à chaque tour
    print(f"Position {position} : {compte}")

print("---")

# --- Méthode 3 : boucler par index avec range(len(liste)) -- utile si on doit MODIFIER pendant le parcours ---
for i in range(len(comptes)):     # range(len(liste)) : 0, 1, 2... jusqu'à la dernière position
    print(f"Index {i} -> {comptes[i]}")

print("---")

# --- Méthode 4 : la compréhension de liste, pour construire une NOUVELLE liste en une ligne (voir Module 7) ---
comptes_maj = [c.upper() for c in comptes]     # .upper() appliqué à chaque compte, résultat collecté dans une nouvelle liste
print(comptes_maj)
```

## 7. List comprehension

La **compréhension de liste** construit une nouvelle liste **en une seule ligne**,
à partir d'une liste existante — c'est un raccourci d'écriture pour un `for` (+ `if`
optionnel) qui remplit une liste au fur et à mesure.

Syntaxe : `[expression for element in iterable if condition]` (le `if` est optionnel).

```python
soldes = [15234.50, -500.0, 8900.0, 0.0, -120.0, 32100.0]

# --- Équivalent "boucle classique" (pour comparaison) ---
soldes_positifs_boucle = []                # liste vide
for s in soldes:                            # parcourt chaque solde
    if s > 0:                                # condition : garder seulement les positifs
        soldes_positifs_boucle.append(s)      # ajoute à la liste résultat
print(soldes_positifs_boucle)

# --- La MÊME chose en compréhension de liste : une seule ligne ---
soldes_positifs = [s for s in soldes if s > 0]     # "s" pour chaque "s" dans soldes, SI s > 0
print(soldes_positifs)

# --- Avec une transformation (pas juste un filtre) : calculer un intérêt de 2% ---
interets = [round(s * 0.02, 2) for s in soldes if s > 0]   # expression = round(s*0.02, 2), condition = s>0
print(interets)

# --- Avec if/else DANS l'expression (attention à l'ordre, différent du if de filtre !) ---
etiquettes = ["POSITIF" if s > 0 else "NEGATIF" for s in soldes]   # if/else avant le "for" = choix de valeur, pas un filtre
print(etiquettes)
```

## 8. Trier une liste (Sort Lists)

- `.sort()` : trie la liste **EN PLACE** (modifie l'originale, ne renvoie rien -> `None`)
- `sorted(liste)` : fonction native qui renvoie une **NOUVELLE** liste triée, sans toucher l'originale
- `reverse=True` : tri décroissant (paramètre commun aux deux)
- `key=` : personnalise le critère de tri (ex : trier par longueur, par valeur absolue...)

```python
soldes = [15234.50, -500.0, 8900.0, 0.0, 32100.0]

soldes.sort()                     # .sort() : trie LA LISTE ELLE-MÊME (en place), croissant par défaut
print("Après .sort() :", soldes)   # soldes a été MODIFIÉ définitivement

soldes.sort(reverse=True)          # reverse=True : tri décroissant
print("Décroissant :", soldes)

# --- sorted() : NE modifie PAS l'original, renvoie une copie triée ---
comptes = ["AC003", "AC001", "AC002"]
comptes_tries = sorted(comptes)     # sorted(liste) : nouvelle liste triée
print("Original (inchangé) :", comptes)
print("Copie triée :", comptes_tries)

# --- key= : trier par un critère personnalisé ---
soldes_bruts = [15234.50, -32000.0, 8900.0, -500.0]
tri_par_montant_absolu = sorted(soldes_bruts, key=abs)   # key=abs : trie en comparant abs(valeur), pas la valeur elle-même
print("Trié par montant absolu :", tri_par_montant_absolu)
```

## 9. Copier une liste (Copy Lists)

**Piège fréquent, identique à NumPy (`Cours_NumPy.ipynb`, Module 7)** : `liste2 = liste1`
ne crée PAS de copie — les deux noms pointent vers **la même liste en mémoire**. Modifier
l'une modifie l'autre. Pour une vraie copie indépendante, utiliser `.copy()`, `list()`
ou le slicing complet `[:]`.

```python
comptes = ["AC001", "AC002", "AC003"]

# --- PIÈGE : ceci n'est PAS une copie, juste un 2e nom pour la MÊME liste ---
alias = comptes                 # alias ET comptes pointent vers le même objet en mémoire
alias.append("AC004")            # on modifie alias...
print("comptes après modif de 'alias' :", comptes)   # ...et comptes a changé aussi !

# --- Vraie copie : 3 façons équivalentes ---
comptes = ["AC001", "AC002", "AC003"]      # réinitialisation
copie1 = comptes.copy()          # .copy() : méthode dédiée, la plus explicite
copie2 = list(comptes)            # list(liste) : la fonction native list() copie aussi
copie3 = comptes[:]                # [:] : slicing complet (du début à la fin) = raccourci historique de copie

copie1.append("AC999")            # on modifie SEULEMENT copie1
print("comptes (inchangé) :", comptes)
print("copie1 (modifiée) :", copie1)
```

## 10. Joindre des listes (Join Lists)

Trois façons de fusionner deux listes en une seule :

## 12. Qu'est-ce qu'un dictionnaire ?

Un **dictionnaire** Python stocke des données sous forme de paires **clé : valeur**.
- **modifiable** (*mutable*)
- **ordonné** (depuis Python 3.7, l'ordre d'insertion est conservé)
- les **clés sont uniques** (pas de doublon de clé — une nouvelle valeur sur une clé
  existante ÉCRASE l'ancienne)

On l'écrit entre accolades `{ }` : `{clé1: valeur1, clé2: valeur2}`.

**Pourquoi c'est essentiel pour Beobank ?** Un dictionnaire représente naturellement
**UN enregistrement** (une ligne d'un fichier CSV, un client, un compte) : chaque
colonne devient une clé, chaque valeur de la ligne devient la valeur associée.
C'est exactement ce que fait `csv.DictReader` (vu au Jour 1) — chaque ligne lue
devient un dictionnaire.

```python
# dict() : fonction native pour créer un dictionnaire vide, équivalent à {}
vide = {}
print(vide, type(vide))     # <class 'dict'>

# Un dictionnaire représentant UN compte Beobank (comme une ligne de CTR.csv)
compte = {
    "IDT_AC": "AC00001",       # clé "IDT_AC" -> valeur "AC00001"
    "SLD_CTR": 15234.50,
    "COD_DEV": "EUR",
    "COD_ECV_CTR": "1",
}
print(compte)

# len() : nombre de PAIRES clé:valeur
print("Nombre de champs :", len(compte))

# "in" : teste si une CLÉ existe (PAS une valeur !)
print("SLD_CTR" in compte)       # True -> "SLD_CTR" est une clé
print(15234.50 in compte)         # False -> 15234.50 est une VALEUR, pas une clé
```

## 16. Supprimer des éléments (Remove Items)

| Méthode / instruction | Supprime | Comportement si clé absente |
|---|---|---|
| `.pop("clé")` | la clé donnée (et RENVOIE sa valeur) | erreur, sauf si un défaut est fourni |
| `.popitem()` | la **dernière** paire clé:valeur insérée (et la RENVOIE en tuple) | erreur si dico vide |
| `del dico["clé"]` | la clé donnée (instruction, ne renvoie rien) | erreur |
| `.clear()` | TOUTES les paires — le dico devient vide `{}` | — |

```python
compte = {"IDT_AC": "AC00001", "SLD_CTR": 15234.50, "COD_DEV": "EUR", "COD_ECV_CTR": "1"}

valeur_retiree = compte.pop("COD_ECV_CTR")     # .pop("clé") : supprime ET renvoie la valeur associée
print("Retiré :", valeur_retiree, "| Reste :", compte)
```

```python

derniere_paire = compte.popitem()               # .popitem() : retire la DERNIÈRE paire insérée, renvoyée en tuple (clé, valeur)
print("Dernière paire retirée :", derniere_paire, "| Reste :", compte)
```

```python

del compte["SLD_CTR"]                            # del dico["clé"] : instruction, supprime sans rien renvoyer
print("Après del :", compte)
```

```python

compte.clear()                                    # .clear() : vide complètement le dictionnaire
print("Après clear() :", compte)
```

## 17. Parcourir un dictionnaire (Loop Dictionaries)

Par défaut, boucler sur un dictionnaire parcourt ses **clés**. Pour les valeurs ou les
paires, on utilise explicitement `.values()` ou `.items()`.

```python
compte = {"IDT_AC": "AC00001", "SLD_CTR": 15234.50, "COD_DEV": "EUR"}

# --- Boucle par défaut : parcourt les CLÉS ---
for cle in compte:                    # équivalent à "for cle in compte.keys()"
    print(cle)

print("---")
```

```python

# --- .keys() : explicite, même résultat que ci-dessus (plus lisible pour un lecteur du code) ---
for cle in compte.keys():             # .keys() : vue sur toutes les clés
    print(cle, "->", compte[cle])      # on peut retrouver la valeur avec compte[cle]

print("---")

```

```python
# --- .values() : parcourt directement les VALEURS, sans les clés ---
for valeur in compte.values():        # .values() : vue sur toutes les valeurs
    print(valeur)

print("---")

```

```python
# --- .items() : LA méthode la plus utilisée -> donne (clé, valeur) en même temps ---
for cle, valeur in compte.items():    # .items() : vue sur les paires (clé, valeur)
    print(f"{cle} = {valeur}")
```

## 21. Qu'est-ce qu'un tuple ?

Un **tuple** ressemble à une liste, avec UNE différence fondamentale :
- **ordonné**, comme une liste
- **autorise les doublons**, comme une liste
- **IMMUABLE** (*immutable*) : une fois créé, on ne peut NI ajouter, NI retirer,
  NI modifier ses éléments

On l'écrit entre parenthèses `( )` (ou même sans rien, juste des virgules).

**Pourquoi utiliser un tuple plutôt qu'une liste ?** Quand les données ne DOIVENT
PAS changer — par exemple les coordonnées d'une agence Beobank (elles ne bougent
pas pendant l'exécution du programme), ou un enregistrement dont on veut garantir
qu'il reste intact une fois créé. Un tuple est aussi légèrement plus rapide et
signale clairement l'intention "ceci est fixe" à qui relit le code.

```python
# tuple() : fonction native pour créer un tuple vide, équivalent à ()
vide = tuple()
print(vide, type(vide))     # <class 'tuple'>

# Une agence Beobank : (nom, latitude, longitude) -- ça ne doit jamais changer une fois défini
agence = ("Bruxelles Centre", 50.8503, 4.3517)
print(agence)

# ATTENTION : un tuple à un seul élément a besoin d'une VIRGULE après la valeur
pas_un_tuple = ("Bruxelles")        # ceci est juste une chaîne de caractères entre parenthèses !
vrai_tuple_un_element = ("Bruxelles",)   # la virgule fait toute la différence
print(type(pas_un_tuple), type(vrai_tuple_un_element))

# len() et "in" fonctionnent comme pour les listes
print(len(agence))
print("Bruxelles Centre" in agence)
```

## 24. Unpacking — déballage de tuple (Unpack Tuples)

Le **unpacking** affecte chaque élément d'un tuple (ou d'une liste) à une variable,
en une seule ligne — très utilisé pour lire des données structurées de façon lisible.

```python
agence = ("Bruxelles Centre", 50.8503, 4.3517)

# Unpacking simple : autant de variables à gauche que d'éléments dans le tuple
ville, latitude, longitude = agence      # déballe les 3 éléments dans 3 variables, dans l'ordre
print(f"{ville} est situee a ({latitude}, {longitude})")
```

```python

# Unpacking AVEC * : récupère "le reste" dans une liste, quand on ne connaît pas le nombre exact
premier, *autres = ("AC001", "AC002", "AC003", "AC004")   # premier = 1er élément, autres = LISTE du reste
print("Premier :", premier, "| Autres :", autres)

premier, *milieu, dernier = ("AC001", "AC002", "AC003", "AC004", "AC005")   # * peut être au milieu
print("Premier :", premier, "| Milieu :", milieu, "| Dernier :", dernier)

# Ignorer une valeur avec "_" (convention : on s'en fiche, mais il faut bien une variable)
_, lat, _ = agence
print("Latitude seule :", lat)
```

## 25. Parcourir un tuple (Loop Tuples)

Identique aux listes : `for`, ou `enumerate()` si on a besoin de la position.
Cas très fréquent : une **liste de tuples**, où chaque tuple est un enregistrement.

```python
portefeuille = [                    # une LISTE de TUPLES : chaque tuple = un compte figé (idt, solde, devise)
    ("AC00001", 15234.50, "EUR"),
    ("AC00002",  -500.00, "EUR"),
    ("AC00003",  8900.00, "USD"),
]

# for + unpacking directement dans la boucle : très lisible
for idt_ac, solde, devise in portefeuille:      # déballage automatique, à chaque tour de boucle
    print(f"{idt_ac} : {solde} {devise}")

print("---")

# Avec enumerate() si on veut aussi la position
for position, (idt_ac, solde, devise) in enumerate(portefeuille, start=1):   # tuple imbriqué à déballer
    print(f"{position}. {idt_ac} -> {solde} {devise}")
```

## 26. Joindre des tuples (Join Tuples)

Comme pour les listes : `+` concatène, `*` répète — les deux renvoient un
**NOUVEAU** tuple (impossible de modifier l'original, immuabilité oblige).

```python
t1 = ("AC001", "AC002")
t2 = ("AC003", "AC004")

fusion = t1 + t2           # + : concatène deux tuples en un nouveau tuple
print(fusion)

repete = ("EUR",) * 3       # * : répète le tuple N fois (virgule obligatoire pour le tuple à 1 élément)
print(repete)
```

## 28. Liste vs Tuple vs Dictionnaire : que choisir ?

| | `list` | `tuple` | `dict` |
|---|---|---|---|
| Écriture | `[1, 2, 3]` | `(1, 2, 3)` | `{"a": 1}` |
| Modifiable ? | ✅ oui | ❌ non (immuable) | ✅ oui |
| Accès | par position (index) | par position (index) | par **clé** |
| Doublons | ✅ autorisés | ✅ autorisés | ❌ clés uniques |
| Cas d'usage type Beobank | une liste de soldes, de comptes | une coordonnée GPS figée, un enregistrement fixe | UN enregistrement (une ligne CSV), un mapping code→libellé |

**Règle simple :** si les données doivent pouvoir évoluer pendant l'exécution
(ajouter/retirer/modifier), utilisez une **liste**. Si elles représentent un
enregistrement figé et structuré, un **tuple**. Si vous avez besoin de retrouver une
valeur par un NOM plutôt qu'une position, un **dictionnaire**.

## 29. Mini-projet final — combiner les 3 structures sur CTR.csv

On charge le vrai fichier `CTR.csv` avec `csv.DictReader` (vu au Jour 1) : chaque
ligne devient un **dictionnaire**, l'ensemble des lignes une **liste de dictionnaires**
— la structure Python la plus naturelle pour représenter un fichier tabulaire.

### 🏁 Exercice final — Rapport de portefeuille (listes + dicts + tuples)

```python
import csv

# TODO 1 : lire ../../data/CTR.csv avec csv.DictReader -> construire une LISTE de DICTIONNAIRES
#          nommée "contrats" (chaque élément = 1 dict = 1 ligne du fichier)
# TODO 2 : construire un dict "libelles" associant chaque code statut à son libellé
#          ("1":"Ouvert", "2":"En attente", "3":"Suspendu", "4":"Clôturé", "5":"En résiliation", "6":"Résilié")
# TODO 3 : avec dict.fromkeys() sur les clés de "libelles", construire un compteur à 0
# TODO 4 : boucler sur "contrats" et incrémenter le compteur du statut de chaque contrat
# TODO 5 : construire une LISTE DE TUPLES (code, libelle, nombre) à partir du compteur et libelles
# TODO 6 : trier cette liste de tuples par nombre décroissant avec sorted(key=..., reverse=True)
#          (key=lambda t: t[2] : trie selon le 3e élément de chaque tuple)
# TODO 7 : afficher le rapport, un tuple par ligne

# --- votre code ---
```

### ✅ Correction Exercice final

```python
import csv

# 1. Lecture -> liste de dictionnaires
contrats = []                                              # liste vide qui va recevoir un dict par ligne
with open("../../data/CTR.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f, delimiter=";")               # chaque ligne lue devient un dict {colonne: valeur}
    for ligne in lecteur:
        contrats.append(ligne)                                # ligne EST DÉJÀ un dict -> on l'ajoute tel quel
print(f"Contrats chargés : {len(contrats)}")

# 2. Dictionnaire de libellés (mapping code -> texte)
libelles = {
    "1": "Ouvert", "2": "En attente", "3": "Suspendu",
    "4": "Clôturé", "5": "En résiliation", "6": "Résilié",
}

# 3. Compteur initialisé à 0 pour chaque code connu
compteur = dict.fromkeys(libelles.keys(), 0)                # dict.fromkeys() : même clés que libelles, valeur 0 partout

# 4. Comptage : un dictionnaire est accédé par sa clé, comme toute ligne CSV
for contrat in contrats:                                     # contrat est un dict (une ligne)
    code = contrat["COD_ECV_CTR"]                              # accès par clé -> le code statut de CETTE ligne
    if code in compteur:                                        # sécurité : ignore un code inattendu
        compteur[code] += 1                                       # incrémente le compteur de ce code

# 5. Liste de tuples (code, libelle, nombre) -- un tuple = un enregistrement figé, idéal pour un export
rapport = [(code, libelles[code], nombre) for code, nombre in compteur.items()]   # list comprehension + unpacking .items()

# 6. Tri par nombre décroissant
rapport_trie = sorted(rapport, key=lambda t: t[2], reverse=True)   # key=lambda t: t[2] -> trie sur le 3e élément du tuple

# 7. Affichage
print("=" * 45)
print("  RAPPORT DE PORTEFEUILLE PAR STATUT")
print("=" * 45)
for code, libelle, nombre in rapport_trie:      # unpacking direct dans le for
    print(f"  {code} ({libelle:15s}) : {nombre:3d} contrats")
print("=" * 45)
```

## 7. Dire à Pandas que c'est un nombre ou une date

Au départ, TOUT est écrit comme du texte dans le fichier, même les nombres et les
dates. Il faut le dire explicitement à Pandas. **On fait ça maintenant, AVANT de
filtrer sur des nombres**, sinon Pandas ne peut pas comparer du texte à un nombre.

````python
# pd.to_numeric(colonne, errors="coerce") : transforme le texte en vrai nombre
# errors="coerce" : si ce n'est pas un nombre, on met une case vide au lieu de planter
tirelires["argent_dedans"] = pd.to_numeric(tirelires["argent_dedans"], errors="coerce")

tirelires["argent_dedans"].dtype   # maintenant c'est bien un nombre (float64)
````

````python
# pd.to_datetime(colonne, format=...) : transforme le texte en vraie date
# format="%Y-%m-%d" : les dates sont écrites "année-mois-jour"
tirelires["date_ouverture"] = pd.to_datetime(tirelires["date_ouverture"], format="%Y-%m-%d", errors="coerce")
tirelires["date_fermeture"] = pd.to_datetime(tirelires["date_fermeture"], format="%Y-%m-%d", errors="coerce")

tirelires["date_ouverture"].head()
````

````python
# .map({...}) : remplace chaque code par un mot qu'on comprend (4 = ouverte, 6 = fermée)
tirelires["statut"] = tirelires["code_statut"].map({4: "Ouverte", 6: "Fermée"})
tirelires["statut"].value_counts()
````

````python
# 1. Transformer toutes les colonnes de dates
colonnes_dates = [
    "date_ouverture",
    "date_fermeture",
    "DAT_ECV_CTR",
    "DAT_MAJ_SLD",
]

for colonne in colonnes_dates:
    if colonne in tirelires.columns:
        tirelires[colonne] = pd.to_datetime(
            tirelires[colonne],
            errors="coerce"
        )

# 2. Extraire des informations de la date d'ouverture
tirelires["annee_ouverture"] = tirelires["date_ouverture"].dt.year
tirelires["mois_ouverture"] = tirelires["date_ouverture"].dt.month
tirelires["nom_mois_ouverture"] = tirelires["date_ouverture"].dt.month_name()
tirelires["trimestre_ouverture"] = tirelires["date_ouverture"].dt.quarter
tirelires["jour_semaine_ouverture"] = tirelires["date_ouverture"].dt.day_name()
tirelires["jour_du_mois"] = tirelires["date_ouverture"].dt.day

# 3. Savoir si la tirelire est fermée
tirelires["est_fermee"] = tirelires["date_fermeture"].notna()

# 4. Durée entre l'ouverture et la fermeture
tirelires["duree_jours"] = (
    tirelires["date_fermeture"] - tirelires["date_ouverture"]
).dt.days

tirelires["duree_annees"] = tirelires["duree_jours"] / 365.25

# 5. Durée jusqu'à aujourd'hui pour les tirelires encore ouvertes
aujourdhui = pd.Timestamp.today().normalize()

tirelires["jours_depuis_ouverture"] = (
    aujourdhui - tirelires["date_ouverture"]
).dt.days

# 6. Début et fin du mois d'ouverture
tirelires["debut_mois_ouverture"] = (
    tirelires["date_ouverture"].dt.to_period("M").dt.start_time
)

tirelires["fin_mois_ouverture"] = (
    tirelires["date_ouverture"].dt.to_period("M").dt.end_time
)

# 7. Filtrer entre deux dates
date_debut = pd.Timestamp("2024-01-01")
date_fin = pd.Timestamp("2025-12-31")

tirelires_ouvertes_periode = tirelires[
    tirelires["date_ouverture"].between(date_debut, date_fin)
]

# 8. Filtrer les tirelires ouvertes aujourd'hui
tirelires_ouvertes = tirelires[
    tirelires["date_fermeture"].isna()
]

# 9. Trier par date
tirelires_triees_par_date = tirelires.sort_values(
    "date_ouverture"
)

# 10. Compter les ouvertures par année
ouvertures_par_annee = (
    tirelires.groupby("annee_ouverture")
    .size()
    .sort_index()
)

# 11. Compter les ouvertures par mois
ouvertures_par_mois = (
    tirelires["date_ouverture"]
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

# 12. Trouver la plus ancienne et la plus récente tirelire
date_plus_ancienne = tirelires["date_ouverture"].min()
date_plus_recente = tirelires["date_ouverture"].max()

print("Première ouverture :", date_plus_ancienne)
print("Dernière ouverture :", date_plus_recente)

# Afficher un résumé
tirelires[
    [
        "numero_tirelire",
        "date_ouverture",
        "date_fermeture",
        "annee_ouverture",
        "nom_mois_ouverture",
        "duree_jours",
        "est_fermee",
    ]
].head()
````

````python
# & veut dire "ET" : les DEUX conditions doivent être vraies
tirelires_riches_en_euros = tirelires[(tirelires["monnaie"] == "EUR") & (tirelires["argent_dedans"] > 1000)]
tirelires_riches_en_euros.head()
````

## 10. Créer de nouvelles colonnes

On peut fabriquer de nouvelles informations à partir de celles qu'on a déjà.

````python
# np.where(condition, si_vrai, si_faux) : un "SI... SINON..." qui marche sur tout le tableau d'un coup
tirelires["en_dette"] = np.where(tirelires["argent_dedans"] < 0, "Oui", "Non")
tirelires["en_dette"].value_counts()
````

````python
# pd.cut(colonne, bins=..., labels=...) : range les nombres dans des "boîtes" (des tranches)
bornes = [-1_000_000, 0, 1000, 10000, 1_000_000]
noms_des_boites = ["En dette", "Petite tirelire", "Tirelire moyenne", "Grosse tirelire"]
tirelires["taille_tirelire"] = pd.cut(tirelires["argent_dedans"], bins=bornes, labels=noms_des_boites)
tirelires["taille_tirelire"].value_counts()
````

````python
# .apply(une_fonction) : applique notre propre fonction à chaque valeur d'une colonne
def dire_bravo_ou_dommage(argent):
    if pd.isna(argent):
        return "On ne sait pas"
    elif argent >= 0:
        return "Bravo !"
    else:
        return "Dommage..."

tirelires["message"] = tirelires["argent_dedans"].apply(dire_bravo_ou_dommage)
tirelires["message"].value_counts()
````

````python
# Définir une fonction qui transforme un montant en message
def dire_bravo_ou_dommage(argent):
    # Vérifier si le montant est manquant
    if pd.isna(argent):
        # Message affiché lorsque le montant est inconnu
        return "On ne sait pas"

    # Vérifier si le montant est positif ou égal à zéro
    elif argent >= 0:
        # Message affiché pour un montant positif ou nul
        return "Bravo !"

    # Si le montant est négatif
    else:
        # Message affiché lorsqu'il y a une dette
        return "Dommage..."


# Appliquer la fonction à chaque valeur de la colonne "argent_dedans"
# Le résultat est enregistré dans une nouvelle colonne appelée "message"
tirelires["message"] = tirelires["argent_dedans"].apply(
    dire_bravo_ou_dommage
)

# Compter le nombre de tirelires pour chaque message
tirelires["message"].value_counts()
````

## 12. Jouer avec les dates

Une fois qu'une colonne est une vraie date, on peut en sortir plein d'informations
avec `.dt`.

````python
# Extraire encore plus d'informations à partir des colonnes de dates déjà transformées

# Semaine de l'année et jour de l'année
tirelires["semaine_ouverture"] = tirelires["date_ouverture"].dt.isocalendar().week
tirelires["jour_annee_ouverture"] = tirelires["date_ouverture"].dt.dayofyear

# Est-ce un jour de week-end ?
tirelires["est_weekend_ouverture"] = tirelires["date_ouverture"].dt.dayofweek >= 5

# Est-ce le début ou la fin du mois ?
tirelires["est_debut_mois"] = tirelires["date_ouverture"].dt.is_month_start
tirelires["est_fin_mois"] = tirelires["date_ouverture"].dt.is_month_end

# Est-ce le début ou la fin de l'année ?
tirelires["est_debut_annee"] = tirelires["date_ouverture"].dt.is_year_start
tirelires["est_fin_annee"] = tirelires["date_ouverture"].dt.is_year_end

# Année bissextile ?
tirelires["annee_bissextile"] = tirelires["date_ouverture"].dt.is_leap_year

# Nombre de jours dans le mois d'ouverture
tirelires["jours_dans_le_mois"] = tirelires["date_ouverture"].dt.days_in_month

# Comparer la date d'ouverture à une date précise avec des décalages (DateOffset)
tirelires["date_ouverture_plus_1an"] = tirelires["date_ouverture"] + pd.DateOffset(years=1)
tirelires["date_ouverture_moins_1mois"] = tirelires["date_ouverture"] - pd.DateOffset(months=1)

# Différence en mois complets entre ouverture et fermeture (ou aujourd'hui si toujours ouverte)
date_reference = tirelires["date_fermeture"].fillna(aujourdhui)
tirelires["duree_mois"] = (
    (date_reference.dt.year - tirelires["date_ouverture"].dt.year) * 12
    + (date_reference.dt.month - tirelires["date_ouverture"].dt.month)
)

# Regrouper par trimestre d'ouverture
ouvertures_par_trimestre = (
    tirelires["date_ouverture"]
    .dt.to_period("Q")
    .value_counts()
    .sort_index()
)

# Regrouper par semaine ISO
ouvertures_par_semaine = (
    tirelires.groupby("semaine_ouverture")
    .size()
    .sort_index()
)

# Utiliser un DatetimeIndex pour rééchantillonner (resample) le nombre d'ouvertures par mois
tirelires_par_date = tirelires.set_index("date_ouverture").sort_index()
ouvertures_resample_mensuel = tirelires_par_date.resample("ME").size()

# Décalage de dates (shift) : comparer chaque ouverture à la précédente (dans l'ordre chronologique)
tirelires_triees_par_date_2 = tirelires.sort_values("date_ouverture").copy()
tirelires_triees_par_date_2["date_ouverture_precedente"] = tirelires_triees_par_date_2["date_ouverture"].shift(1)
tirelires_triees_par_date_2["jours_depuis_ouverture_precedente"] = (
    tirelires_triees_par_date_2["date_ouverture"] - tirelires_triees_par_date_2["date_ouverture_precedente"]
).dt.days

# Aperçu de toutes ces nouvelles colonnes liées aux dates
tirelires[
    [
        "numero_tirelire",
        "date_ouverture",
        "semaine_ouverture",
        "jour_annee_ouverture",
        "est_weekend_ouverture",
        "est_debut_mois",
        "est_fin_mois",
        "annee_bissextile",
        "jours_dans_le_mois",
        "duree_mois",
    ]
].head(10)
````

````python
tirelires["annee_ouverture"] = tirelires["date_ouverture"].dt.year    # .dt.year : l'année de la date
tirelires["mois_ouverture"] = tirelires["date_ouverture"].dt.month    # .dt.month : le mois (1 à 12)
tirelires[["date_ouverture", "annee_ouverture", "mois_ouverture"]].head()
````

````python
# on calcule combien de jours se sont écoulés depuis l'ouverture de chaque tirelire
aujourdhui = pd.Timestamp.today()   # la date d'aujourd'hui
tirelires["jours_depuis_ouverture"] = (aujourdhui - tirelires["date_ouverture"]).dt.days
tirelires[["date_ouverture", "jours_depuis_ouverture"]].head()
````

````python
# Filtrer par ANNÉE : garder seulement les tirelires ouvertes en 2025
tirelires_2025 = tirelires[tirelires["date_ouverture"].dt.year == 2025]
print(len(tirelires_2025), "tirelires ouvertes en 2025")

# Filtrer par MOIS : garder seulement les tirelires ouvertes en janvier (mois 1)
tirelires_janvier = tirelires[tirelires["date_ouverture"].dt.month == 1]
print(len(tirelires_janvier), "tirelires ouvertes en janvier")

# Filtrer par JOUR : garder seulement les tirelires ouvertes un lundi (0 = lundi, 6 = dimanche)
tirelires_lundi = tirelires[tirelires["date_ouverture"].dt.dayofweek == 0]
print(len(tirelires_lundi), "tirelires ouvertes un lundi")

# On peut combiner avec & : année 2025 ET mois de janvier
tirelires_janvier_2025 = tirelires[
    (tirelires["date_ouverture"].dt.year == 2025) &
    (tirelires["date_ouverture"].dt.month == 1)
]

# .between() : garder les tirelires ouvertes ENTRE deux dates
tirelires_entre_deux_dates = tirelires[
    tirelires["date_ouverture"].between("2025-01-01", "2025-12-31")
]

# .isin([...]) : garder les tirelires ouvertes dans l'une de ces années
tirelires_annees_recentes = tirelires[
    tirelires["date_ouverture"].dt.year.isin([2024, 2025, 2026])
]

# .dt.to_period("M") : comparer directement un mois entier (ex : "2025-01")
tirelires_mois_precis = tirelires[
    tirelires["date_ouverture"].dt.to_period("M") == "2025-01"
]

tirelires_janvier_2025[["numero_tirelire", "date_ouverture"]].head()
````

## 16. Un tableau qui croise deux informations

## 22. Pourquoi Pandas est rapide

Pandas travaille sur TOUTE une colonne d'un coup (ça s'appelle "vectoriser"), au lieu
de regarder les tirelires une par une avec une boucle. C'est beaucoup plus rapide.

````python
import time

depart = time.perf_counter()
resultat_boucle = []
for valeur in tirelires["argent_dedans"]:                 # une boucle : on regarde chaque tirelire une par une
    resultat_boucle.append(valeur * 2 if pd.notna(valeur) else None)
duree_boucle = time.perf_counter() - depart

depart = time.perf_counter()
resultat_direct = tirelires["argent_dedans"] * 2           # vectorisé : toute la colonne d'un coup
duree_directe = time.perf_counter() - depart

print(f"Avec une boucle : {duree_boucle*1000:.3f} ms")
print(f"Vectorisé       : {duree_directe*1000:.3f} ms")
````

## 23. Tout refaire d'un coup avec une fonction

Une fois qu'on connaît toutes les étapes, on peut les ranger dans une seule fonction,
pour ne plus jamais avoir à les retaper.

````python
def ouvrir_et_nettoyer_les_tirelires():
    """Ouvre CTR.csv et le nettoie, en une seule fois."""
    df = pd.read_csv("../../data/CTR.csv", sep=";", encoding="cp1252")   # 1. on ouvre le fichier
    df = df.replace(".", pd.NA)                                        # 2. les points deviennent des cases vides
    df["SLD_CTR"] = pd.to_numeric(df["SLD_CTR"], errors="coerce")      # 3. le solde devient un vrai nombre
    df["DAT_OUV_CTR"] = pd.to_datetime(df["DAT_OUV_CTR"], format="%Y-%m-%d", errors="coerce")   # 4. vraie date
    df["statut"] = df["COD_ECV_CTR"].map({4: "Ouverte", 6: "Fermée"})  # 5. un mot au lieu d'un code
    return df

tirelires_propres = ouvrir_et_nettoyer_les_tirelires()
tirelires_propres.info()
````

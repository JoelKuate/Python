## Module 1 — Algorithmes et introduction à Python
### 1.1 Qu'est-ce qu'un algorithme ?

Un **algorithme** est une suite d'étapes précises et ordonnées qui résout un problème.
Avant d'écrire une seule ligne de code, on peut décrire un algorithme en français :

1. Lire le solde du compte
2. Lire le taux d'intérêt annuel
3. Diviser le taux par 12 pour obtenir le taux mensuel
4. Multiplier le solde par ce taux mensuel
5. Afficher le résultat

Un langage de programmation (ici Python) sert à **traduire** cet algorithme en instructions
qu'un ordinateur peut exécuter, sans ambiguïté.

### 1.2 Pourquoi Python ?

Python est très utilisé en entreprise, y compris dans la finance, pour plusieurs raisons :

- **Lisible** : la syntaxe ressemble à de l'anglais courant, peu de symboles cryptiques
- **Gratuit et open source** : pas de licence à payer
- **Écosystème immense** : pandas (données), numpy (calcul), matplotlib (graphiques)...
- **Polyvalent** : scripts, analyse de données, applications web, automatisation
- **Communauté énorme** : documentation et solutions faciles à trouver

C'est pour ces raisons que Beobank fait migrer ses traitements vers Python.

```python
# ── F-strings — interpolation de variables dans du texte ──────────────────
# En Python : f"Compte {idt_ac} : {solde:.2f}"
# Le f devant les guillemets active l'interpolation

idt_ac  = "AC00001"
devise  = "EUR"
solde   = 15234.5
statut  = "Ouvert"

# F-string simple — les {} contiennent les variables
msg1 = f"Compte : {idt_ac}"
print(msg1)   # Compte : AC00001

# F-string avec format numérique — :.2f = 2 décimales
msg2 = f"Solde : {solde:.2f} {devise}"
print(msg2)   # Solde : 15234.50 EUR


# F-string complet — rapport d'une ligne
msg3 = f"[{idt_ac}] Statut={statut} | Solde={solde:,.2f} {devise}"
print(msg3)   # [AC00001] Statut=Ouvert | Solde=15,234.50 EUR
```

## Module 3 — Opérateurs
### 3.1 Opérateurs arithmétiques

```python
# ── Opérateurs arithmétiques ───────────────────────────────────────────────
a, b = 17, 5

print(a + b)    # addition        → 22
print(a - b)    # soustraction    → 12
print(a * b)    # multiplication  → 85
print(a / b)    # division réelle → 3.4  (toujours un float)
print(a // b)   # division entière (quotient) → 3
print(a % b)    # modulo (reste de la division) → 2
print(a ** b)   # puissance → 17 puissance 5

# Cas d'usage Beobank : convertir un nombre de mois en années + mois restants
nb_mois = 26
annees        = nb_mois // 12     # division entière → 2
mois_restants = nb_mois % 12      # reste           → 2
print(f"{nb_mois} mois = {annees} an(s) et {mois_restants} mois")
```

### 3.2 Opérateurs de comparaison

```python
# ── Opérateurs de comparaison — renvoient toujours True ou False ─────────
solde = 250.0

print(solde == 250.0)   # égalité           → True
print(solde != 0)       # différence        → True
print(solde > 100)      # supérieur         → True
print(solde >= 250.0)   # supérieur ou égal → True
print(solde < 0)        # inférieur         → False
print(solde <= 0)       # inférieur ou égal → False

# Piège fréquent : = (affectation) est différent de == (comparaison)
# solde = 0    → affecte 0 à solde
# solde == 0   → teste si solde vaut 0 (ne modifie rien)
```

### 3.3 Opérateurs logiques — and, or, not

```python
# ── Opérateurs logiques — combiner plusieurs conditions ───────────────────
solde  = -150.0
statut = "1"     # "1" = compte ouvert

# and : VRAI seulement si LES DEUX conditions sont vraies
compte_a_risque = statut == "1" and solde < 0
print(compte_a_risque)   # True

# or : VRAI si AU MOINS UNE condition est vraie
a_verifier = solde < 0 or solde > 100000
print(a_verifier)        # True

# not : inverse une condition (True devient False et inversement)
compte_ferme = not (statut == "1")
print(compte_ferme)      # False

# Table de vérité rapide pour and / or
a = 15
for x in [True, False]:
    for y in [True, False]:
        print(f"  {str(x):{a}s} and {str(y):{a}s} = {str(x and y):{a}s}  |  or = {x or y}")
```

### 3.4 Assignation augmentée

```python
# ── Assignation augmentée — raccourcis pour "modifier puis réaffecter" ───
solde = 1000.0

solde += 250      # équivalent de : solde = solde + 250
print(solde)      # 1250.0

solde -= 100      # équivalent de : solde = solde - 100
print(solde)      # 1150.0

solde *= 1.035    # équivalent de : solde = solde * 1.035  (appliquer 3.5% d'intérêt)
print(round(solde, 2))

compteur = 0
for joel in range(5):
    compteur *= 3   # incrémenter le compteur à chaque tour de boucle
    print(f"Compteur après tour {joel} : {compteur}")
print(f"Compteur final : {compteur}")
print(f"Compteur après boucle : {joel}")
```

## Module 4 — Conditions if / elif / else
### 4.1 Structure de base

```python
# ── if / elif / else ──────────────────────────────────────────────────────
# Python : if / elif (= else if) / else — avec INDENTATION (4 espaces)

cod_ecv_ctr = "3"   # code statut du contrat (depuis CTR.csv)

# Chaque elif teste une nouvelle condition
# Python s'arrête dès qu'une condition est vraie
if cod_ecv_ctr == "1":
    libelle = "Ouvert"
elif cod_ecv_ctr == "2":
    libelle = "En attente"
elif cod_ecv_ctr == "3":
    libelle = "Suspendu"
elif cod_ecv_ctr == "4":
    libelle = "Clôturé"
elif cod_ecv_ctr == "5":
    libelle = "En résiliation"
elif cod_ecv_ctr == "6":
    libelle = "Résilié"
else:
    libelle = "Inconnu"  # filet de sécurité si code non reconnu

print(f"Statut {cod_ecv_ctr} = {libelle}")   # Statut 3 = Suspendu
```

### 4.2 Conditions composées

```python
# ── Conditions composées avec and / or / not / in ─────────────────────────

cod_ecv_ctr = "1"   # contrat ouvert
solde       = -250.0

# AND : les DEUX conditions doivent être vraies
# OR  : au moins UNE condition doit être vraie
# Parenthèses pour contrôler l'ordre d'évaluation (comme en algèbre)
if cod_ecv_ctr == "1" and (solde < 0 or solde < 100):
    risque = "CRITIQUE — compte ouvert avec solde anormal"
elif cod_ecv_ctr in ["4", "6"]:    # in [] = tester l'appartenance à une liste
    risque = "Inactif"
else:
    risque = "Normal"

print(risque)   # CRITIQUE — compte ouvert avec solde anormal

# not : inverser une condition
if not cod_ecv_ctr == "4":
    print("Le compte n'est PAS clôturé")
```

## Module 5 — Boucles for, while, range
### 5.1 Boucle for + range

```python
# ── Boucle for — itérer sur une séquence ──────────────────────────────────
# Python : for i in range(1, 13): ...  (1 inclus, 13 EXCLU)

# Simulation d'un rapport mensuel sur 12 mois
print("=== Rapport mensuel Beobank ===")
for mois in range(1, 13):        # range(1,13) génère 1,2,3,...,12
    # :02d = format entier sur 2 chiffres (01, 02, ..., 12)
    print(f"  Traitement mois {mois:02d}/2024...")

# Boucle sur une liste — itérer sur des valeurs connues
devises = ["EUR", "USD", "GBP", "CHF"]
for devise in devises:           # devise prend la valeur de chaque élément
    print(f"  Devise : {devise}")
```

```python
# ── enumerate() — index + valeur simultanément ────────────────────────────
# Python : enumerate() donne automatiquement (position, valeur)

comptes = ["AC00001", "AC00042", "AC00099", "AC00150"]

# enumerate(comptes, start=1) → (1,"AC00001"), (2,"AC00042"), ...
for rang, compte in enumerate(comptes, start=1):
    print(f"  Rang {rang:02d} : {compte}")

# Accumuler un total dans une boucle
total = 0
soldes = [15234, 8900, -500, 32100, 4500]
for s in soldes:
    if s > 0:           # ne compter que les soldes positifs
        total += s      # += est un raccourci pour total = total + s
print(f"Total des soldes positifs : {total} EUR")
```

### 5.2 Boucle while

```python
# ── Boucle while — répéter tant qu'une condition est vraie ────────────────
# Utile quand le nombre d'itérations n'est pas connu à l'avance
# ATTENTION : toujours prévoir une condition de sortie (sinon boucle infinie)

solde      = 1000.0    # solde initial du compte
nb_retraits = 0        # compteur de retraits
montant     = 150.0    # montant de chaque retrait

# Tant que le solde permet un retrait...
while solde >= montant:
    solde -= montant    # déduire le montant (-= raccourci pour solde = solde - montant)
    nb_retraits += 1    # compter ce retrait

print(f"{nb_retraits} retraits effectués")
print(f"Solde restant : {solde:.2f} EUR")   # 100.00 EUR

# break — sortir d'une boucle avant la fin
for i in range(100):
    
    if i == 5:
        break           # sortir immédiatement quand i = 5
    print(f"  i = {i}")
```

```python
# ── continue — passer au tour suivant sans exécuter le reste du bloc ─────
# Différence avec break : continue ne sort PAS de la boucle, il saute juste ce tour

for i in range(10):
    if i % 2 != 0:     
        continue        # saute directement au i suivant (le print ci-dessous est ignoré)
    print(f"  i pair = {i}")
```

## Module 6 — Listes et Compréhensions
### 6.1 Listes

```python
# ── Listes — séquences ordonnées et modifiables ───────────────────────────
# Créées avec des crochets []
# L'index commence à 0 (le premier élément)

devises = ["EUR", "USD", "GBP", "CHF"]   # liste de 4 éléments

# Accès par index
print(devises[0])    # EUR  — premier (index 0)
print(devises[-1])   # CHF  — dernier (index -1 = depuis la fin)
print(devises[1:3])  # ['USD', 'GBP'] — tranche : de l'index 1 inclus à 3 exclu

# Taille
print(len(devises))  # 4

# Modifier la liste
devises.append("JPY")       # ajouter en fin
devises.insert(1, "CAD")    # insérer à la position 1
devises.remove("GBP")       # supprimer la valeur "GBP"
print(devises)

# Trier
devises.sort()              # tri en place (modifie la liste)
print(sorted(devises))      # retourne une copie triée (ne modifie pas)
dir(devises)
```

### 6.2 List comprehensions

```python
# ── List comprehensions ────────────────────────────────────────────────────
# Syntaxe : [expression FOR element IN sequence IF condition]
# Très puissant — remplace un for loop de transformation simple

soldes = [15234, -500, 8900, 0, 32100, -120, 4500, 1200, -85, 9999]

# Extraire seulement les soldes positifs (WHERE solde > 0)
soldes_positifs = [s for s in soldes if s > 0]
print("Positifs :", soldes_positifs)

# Calculer les intérêts sur les soldes positifs
taux_annuel = 0.035
interets = [round(s * taux_annuel / 12, 2) for s in soldes if s > 0]
print("Intérêts :", interets)

# Créer des labels — "OK" ou "NÉGATIF" selon le solde
labels = ["OK" if s >= 0 else "NÉGATIF" for s in soldes]
print("Labels :", labels)

# Compter combien de soldes sont négatifs
nb_negatifs = sum(1 for s in soldes if s < 0)  # sum() d'un générateur
print(f"Soldes négatifs : {nb_negatifs}")
```

### 7.3 Réduction — ramener une séquence à une seule valeur

```python
# ── Réduction — fonctions qui résument une séquence en UNE valeur ────────
soldes = [15234, -500, 8900, 0, 32100]

print(len(soldes))    # nombre d'éléments          → 5
print(sum(soldes))    # somme de tous les éléments → 55734
print(min(soldes))    # plus petite valeur         → -500
print(max(soldes))    # plus grande valeur         → 32100

# any() : True si AU MOINS UN élément est vrai (ici : au moins un solde négatif ?)
print(any(s < 0 for s in soldes))   # True

# all() : True si TOUS les éléments sont vrais (ici : tous les soldes positifs ?)
print(all(s >= 0 for s in soldes))  # False
```

## Module 8 — Tuples et Dictionnaires
### 8.1 Tuples

```python
# ── Tuples — séquences IMMUABLES ──────────────────────────────────────────
# Créés avec des parenthèses ()
# Immuable = impossible de modifier après création
# Utilisation typique : paires de valeurs, coordonnées, retour multi-valeur

# Tuple représentant une agence Beobank (nom, latitude, longitude)
agence = ("Bruxelles Centre", 50.8503, 4.3517)

# Accès par index (comme une liste)
print(agence[0])   # Bruxelles Centre

# Déballage (unpacking) — affecter chaque valeur à une variable nommée
ville, lat, lon = agence                      # déballage complet
print(f"Agence {ville} — coordonnées : {lat:.4f}, {lon:.4f}")

# Tuples dans une liste = table de données simple
portefeuille = [
    ("AC00001", 15234.50, "EUR"),   # (compte, solde, devise)
    ("AC00042", -500.00,  "EUR"),
    ("AC00099", 8900.00,  "USD"),
]

# Déballage dans la boucle for
for idt, solde, dev in portefeuille:
    flag = "OK" if solde >= 0 else "ALERTE"
    print(f"  {idt} : {solde:10.2f} {dev} [{flag}]")
```

## Module 9 — Chaînes de caractères, Sets et Unpacking
### 9.1 Méthodes de chaînes usuelles

```python
# ── Méthodes de chaînes de caractères ─────────────────────────────────────
# Une chaîne (str) a de nombreuses méthodes intégrées.
# Elles NE modifient JAMAIS la chaîne d'origine : chacune retourne une nouvelle chaîne.

identifiant_brut = "  ac00001 \n"          # valeur "sale" telle qu'on peut la recevoir d'un fichier

identifiant_sans_espaces = identifiant_brut.strip()        # .strip() retire espaces/retours à la ligne en début et fin
identifiant_normalise    = identifiant_sans_espaces.upper() # .upper() met tout en majuscules
print(identifiant_normalise)                # AC00001

# .replace(ancien, nouveau) — remplacer une sous-chaîne
libelle_produit = "Compte-Courant-Beobank"
libelle_lisible = libelle_produit.replace("-", " ")
print(libelle_lisible)                      # Compte Courant Beobank

# .split(separateur) — découper une chaîne en liste
ligne_brute = "AC00001;15234.50;EUR;1"
champs = ligne_brute.split(";")             # découpe sur le point-virgule
print(champs)                               # ['AC00001', '15234.50', 'EUR', '1']

# .join(liste) — l'inverse de .split() : assembler une liste en une seule chaîne
ligne_reconstruite = ";".join(champs)
print(ligne_reconstruite)                   # AC00001;15234.50;EUR;1

# .startswith() / .endswith() — tester un préfixe ou un suffixe
print(identifiant_normalise.startswith("AC"))   # True
```

### 9.2 Sets — ensembles de valeurs uniques

```python
# ── Sets — collections non ordonnées de valeurs UNIQUES ───────────────────
# Créés avec des accolades {} (comme un dict, mais sans les ":")
# Un set élimine automatiquement les doublons

statuts_observes = ["1", "1", "3", "4", "1", "6", "2", "1", "4"]

# set() convertit une liste en ensemble de valeurs uniques
statuts_uniques = set(statuts_observes)
print(statuts_uniques)              # {'1', '2', '3', '4', '6'} — ordre non garanti
print(len(statuts_uniques))         # 5 statuts différents observés

# Opérations d'ensembles — utile pour comparer deux populations de comptes
comptes_janvier = {"AC001", "AC002", "AC003"}
comptes_fevrier = {"AC002", "AC003", "AC004"}

print("Communs :",  comptes_janvier & comptes_fevrier)   # intersection : présents les 2 mois
print("Tous :",     comptes_janvier | comptes_fevrier)   # union : tous les comptes des 2 mois
print("Perdus :",   comptes_janvier - comptes_fevrier)   # différence : comptes de janvier absents en février
```

## Module 10 — Fonctions, Lambda et try/except
### 10.1 Définir une fonction

```python
# ── def — définir une fonction réutilisable ────────────────────────────────
# Une fonction = un bloc de code nommé, appelable autant de fois que nécessaire


def classifier_solde(solde, seuil_bas=500, seuil_haut=10000):
    if solde < seuil_bas:        # en dessous du seuil bas
        return "Faible"
    elif solde < seuil_haut:     # entre les deux seuils
        return "Moyen"
    else:                        # au-dessus du seuil haut
        return "Élevé"
# Appels de la fonction avec différentes valeurs
print(classifier_solde(250))         # Faible (250 < 500)
print(classifier_solde(5000))        # Moyen  (500 ≤ 5000 < 10000)
print(classifier_solde(25000))       # Élevé  (25000 ≥ 10000)

# Paramètres nommés — ordre libre
print(classifier_solde(solde=250, seuil_bas=100))   # Moyen (100 ≤ 250 < 10000)
```

```python
# ── Retourner plusieurs valeurs + paramètres par défaut ───────────────────

def calculer_interets(solde, taux=0.035, duree_mois=12):
    """Calcule les intérêts sur un solde.

    Paramètres : solde (float), taux (float, défaut 3.5%), duree_mois (int)
    Retourne   : (interets_annuels, interets_mensuels) — tuple de 2 valeurs
    """
    interets_annuels  = round(solde * taux, 2)           # intérêts sur 1 an
    interets_mensuels = round(solde * taux / 12, 2)      # intérêts sur 1 mois
    return interets_annuels, interets_mensuels            # retourner 2 valeurs

# Déballage du tuple retourné
annuels, mensuels = calculer_interets(10000)
print(f"Taux 3.5% — Annuel: {annuels} EUR | Mensuel: {mensuels} EUR")

# Appel avec taux personnalisé
annuels5, mensuels5 = calculer_interets(10000, taux=0.05)
print(f"Taux 5.0% — Annuel: {annuels5} EUR | Mensuel: {mensuels5} EUR")
```

### 10.3 try / except

```python
# ── try / except — gérer les erreurs sans planter ─────────────────────────
# En Python, on doit gérer explicitement les cas d'erreur

def convertir_solde_robuste(valeur):
    """
    Convertit une valeur texte en float.
    Retourne None si la conversion est impossible (valeur manquante, texte).

    Retourne None si la valeur est manquante ou non convertible.
    """
    try:
        # Tenter la conversion — peut échouer si valeur = ".", "N/A", None...
        return float(valeur)
    except (ValueError, TypeError):
        # ValueError : float("texte") impossible
        # TypeError  : float(None) impossible
        return None   # retourner None = valeur manquante Python

# Tests sur différents types d'entrées Beobank
valeurs_test = ["15234.50", ".", "N/A", None, "8900", "", "1 234.50"]
for v in valeurs_test:
    resultat = convertir_solde_robuste(v)
    print(f"  Entrée: {repr(v):15s} → {resultat}")
```

```python
import builtins

builtins.ValueError 
try:
    a = int(input("Entrez le numérateur : "))
    b = int(input("Entrez le dénominateur : "))
    resultat = a / b
except ZeroDivisionError:
    print("Erreur : Vous ne pouvez pas diviser par zéro.")
except ValueError:
    print("Erreur : Veuillez entrer un nombre entier valide.")
else:
    # Ce bloc s'exécute UNIQUEMENT si aucune exception n'a été levée
    print(f"Le résultat de la division est : {resultat}")
finally:
    # Ce bloc s'exécute TOUJOURS (qu'il y ait eu une erreur ou non)
    print("Fin du calcul.")
```

### Gestion avancée des exceptions

```python
# ── Plusieurs types d'exceptions, finally, et raise ───────────────────────

def diviser(a, b):
    try:
        return a / b
    except ZeroDivisionError:                # erreur précise : division par zéro
        print("Erreur : division par zéro")
        return None
    except TypeError:                        # erreur précise : types incompatibles
        print("Erreur : types incompatibles")
        return None
    finally:
        print("Tentative de division terminée")   # s'exécute TOUJOURS, erreur ou pas

print(diviser(10, 2))    # 5.0
print(diviser(10, 0))    # None (ZeroDivisionError gérée)
print(diviser(10, "a"))  # None (TypeError gérée)

# raise — déclencher volontairement une exception
def verifier_solde(solde):
    if solde < -10000:
        raise ValueError(f"Solde anormal : {solde} (limite -10000)")
    return solde

try:
    verifier_solde(-50000)
except ValueError as erreur:
    print(f"Rejeté : {erreur}")
```

## Module 13 — Lecture CSV avec csv.DictReader
### 13.1 Lire CTR.csv

```python
# ── csv.DictReader — lire un CSV ligne par ligne ──────────────────────────
# Python : csv.DictReader — chaque ligne devient un dictionnaire automatiquement

import csv         # module inclus dans Python, pas besoin de pip

contrats = []      # liste vide qui va accueillir les lignes

# with open() = ouverture propre du fichier (fermé automatiquement à la fin)
with open("../../data/CTR.csv", encoding="utf-8") as f:
    #f = open("../../data/CTR.csv", encoding="utf-8")   # ancienne méthode, moins sûre
    # DictReader lit la première ligne comme en-tête (noms des colonnes)
    # delimiter=";" car les fichiers Beobank utilisent le point-virgule
    lecteur = csv.DictReader(f, delimiter=";") #itérateur qui lit toutes les premières lignes et les stocke comme clé

    # À chaque itération, le lecteur avance d'une ligne dans le fichier, découpe les valeurs et construit un dictionnaire associant chaque colonne à sa valeur textuelle
    for ligne in lecteur:
        contrats.append(ligne)    # ajouter le dict de la ligne à la liste

# Afficher le nombre total de lignes lues dans le fichier CSV
print(f"Nombre de lignes lues : {len(contrats)}")

# Afficher un titre avant de présenter la première ligne
print("\nPremière ligne (dict) :")

# Parcourir les paires clé-valeur de la première ligne
for cle, val in contrats[0].items():
    # Afficher le nom de la colonne et la valeur correspondante
    print(f"  {cle:20s} : {val}")
```

```python
# ── Analyser les données du CSV — équivalent de PROC FREQ ─────────────────

# Créer un dictionnaire vide pour stocker le nombre de contrats par statut.
comptage = {}

# Parcourir chaque contrat contenu dans la liste contrats.
for ctr in contrats:
    # Récupérer le code statut du contrat courant.
    statut = ctr["COD_ECV_CTR"]

    # Vérifier si le statut n'existe pas encore dans le dictionnaire.
    if statut not in comptage:
        # Initialiser le compteur de ce statut à zéro.
        comptage[statut] = 0

    # Ajouter un contrat au compteur du statut.
    comptage[statut] += 1

# Afficher un titre pour présenter la distribution des statuts.
print("=== Distribution des statuts (PROC FREQ équivalent) ===")

# Calculer le nombre total de contrats.
total = len(contrats)

# Parcourir les codes de statut dans l'ordre croissant.
for cod in sorted(comptage.keys()):
    # Récupérer le libellé du statut dans le dictionnaire statuts.
    # Utiliser « Inconnu » si le code n'existe pas.
    lib = statuts.get(cod, "Inconnu")

    # Récupérer le nombre de contrats correspondant au statut.
    n = comptage[cod]

    # Calculer le pourcentage de contrats pour ce statut.
    pct = n / total * 100

    # Afficher le code, le libellé, le nombre et le pourcentage.
    print(f"  {cod} — {lib:20s} : {n:3d} ({pct:5.1f}%)")

# Compter les contrats par statut (comme PROC FREQ DATA=CTR; TABLES COD_ECV_CTR;)
comptage = {}       # dictionnaire pour accumuler les comptages

for ctr in contrats:
    statut = ctr["COD_ECV_CTR"]           # lire le code statut de cette ligne
    if statut not in comptage:            # première fois qu'on voit ce code ?
        comptage[statut] = 0              # initialiser le compteur à 0
    comptage[statut] += 1                 # incrémenter de 1

# Afficher comme PROC FREQ (trié par code)
print("=== Distribution des statuts (PROC FREQ équivalent) ===")
total = len(contrats)
for cod in sorted(comptage.keys()):       # sorted() = tri alphabétique
    lib = statuts.get(cod, "Inconnu")
    n   = comptage[cod]
    pct = n / total * 100
    print(f"  {cod} — {lib:20s} : {n:3d} ({pct:5.1f}%)")


```
